```python
# ============================================================================
# IMPORT LIBRARIES AND USER-DEFINED MODULES
# ============================================================================
%load_ext autoreload
%autoreload 2

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# Lithium thermophysical-property equations
import lithium_properties as li

# Tungsten thermophysical-property equations
import tungsten_properties as tu

# Heat-pipe operating conditions and geometry
import heat_pipe_inputs as hp

# Tungsten-Rhenium alloy thermophysical-property data
import tungsten_rhenium_properties as wre
```

```python
# ============================================================================
# LOAD HEAT-PIPE OPERATING CONDITIONS AND GEOMETRY
# ============================================================================
# Operating conditions
# ------------------------------------------------------------------------
T_range = hp.temperature_range()
T = T_range
n = hp.n_wick_layers

# ------------------------------------------------------------------------
# Heat-pipe geometry
# ------------------------------------------------------------------------
do = hp.do
di = hp.di
Lt = hp.Lt
Le = hp.Le
La = hp.La
Lc = hp.Lc
psi = hp.psi
g = hp.g

# ------------------------------------------------------------------------
# Wick structure
# ------------------------------------------------------------------------
S = hp.S
N = hp.N
dw = hp.dw

# ------------------------------------------------------------------------
# Check imported inputs
# ------------------------------------------------------------------------
print("Heat-pipe inputs loaded successfully.")
print("-------------------------------------")
print(f"Temperature range      : {T[0]:.0f} K to {T[-1]:.0f} K")
print(f"Temperature increment  : {T[1] - T[0]:.0f} K")
print(f"Temperature points     : {len(T)}")
print(f"Wick layers            : {n}")
print(f"Outer diameter         : {do:.3f} m")
print(f"Inner diameter         : {di:.3f} m")
print(f"Total length           : {Lt:.3f} m")
print(f"Evaporator length      : {Le:.3f} m")
print(f"Adiabatic length       : {La:.3f} m")
print(f"Condenser length       : {Lc:.3f} m")
print(f"Gravity                : {g:.2f} m/s²")
```

```python
# ============================================================================
# MAX CAPILLARY PRESSURE CALCULATION
# ============================================================================
# S. W. Chi
# Eq. (2-4): P_cm = 2*sigma/r_c
# For a wire-mesh wick: r_c = (w + d_w)/2 from the table 2-1
# ============================================================================
# Wire-center spacing or mesh pitch [m]
wire_pitch = 1.0 / N\
# Clear spacing between adjacent wires [m]
W_s = wire_pitch - dw

# Effective capillary radius [m]
r_c = (W_s + dw) / 2.0

# Lithium surface tension over the temperature range [N/m]
sigma = li.surface_tension(T)

# Maximum capillary pressure available at each temperature [Pa]
P_cm = (2.0 * sigma) / r_c

# Index of maximum surface tension
max_sigma_index = np.argmax(sigma)

# Maximum values
max_sigma = sigma[max_sigma_index]
T_at_max_sigma = T[max_sigma_index]
max_P_cm = P_cm[max_sigma_index]

# Display results
print("Capillary-pressure calculation completed.")
print("-----------------------------------------")
print(f"Wire pitch                 : {wire_pitch:.8f} m")
print(f"Clear wire spacing         : {W_s:.8f} m")
print(f"Effective capillary radius : {r_c:.8f} m")
print(f"Maximum surface tension    : {max_sigma:.6f} N/m")
print(f"Temperature at maximum     : {T_at_max_sigma:.2f} K")
print(f"Maximum capillary pressure : {max_P_cm:.2f} Pa")
```

```python
# ============================================================================
# LIQUID PRESSURE DROP CALCULATION
# ============================================================================
# S. W. Chi, Eqs. (2-17), (2-19), (2-21), (2-22), (2-27), (2-28)
# ============================================================================
# Force T to the full temperature array — don't rely on it being set
# correctly by an earlier cell in this kernel session.
T = hp.temperature_range()

# ------------------------------------------------------------------------
# Wick / vapor-core geometry
# ------------------------------------------------------------------------
# Vapor-core diameter = pipe i.d. minus twice the total radial wick
# thickness (n layers stacked against the inner wall)
dv = hp.di - 2.0 * (hp.n_wick_layers * hp.T_w)

# Wick cross-sectional area = annulus between pipe i.d. and vapor core
A_w = np.pi * ((hp.di ** 2 - dv ** 2) / 4.0)

# Porosity for wrapped-screen wick, Eq. (2-28): epsilon = 1 - (pi*S*N*d)/4
epsilon = 1.0 - (hp.S * np.pi * hp.N * hp.dw) / 4.0

# Permeability for wrapped-screen wick, Eq. (2-27): K = d^2*eps^3 / [122*(1-eps)^2]
K = (hp.dw ** 2) * (epsilon ** 3) / (122.0 * (1.0 - epsilon) ** 2)

# ------------------------------------------------------------------------
# Liquid properties over the operating-temperature range
# ------------------------------------------------------------------------
rho_l = li.liquid_density(T)
h_fg = li.latent_heat_vaporization(T)
mu = li.liquid_dynamic_viscosity(T)

# Sanity check — catch shape mismatches immediately instead of
# hitting an obscure IndexError three lines later
assert np.shape(T) == np.shape(rho_l) == np.shape(h_fg) == np.shape(mu), \
    f"Shape mismatch: T{np.shape(T)}, rho_l{np.shape(rho_l)}, h_fg{np.shape(h_fg)}, mu{np.shape(mu)}"

# ------------------------------------------------------------------------
# Liquid velocity and frictional coefficient
# ------------------------------------------------------------------------
# Eq. (2-19): V_l = Q / (epsilon * A_w * rho_l * lambda)
Vl = hp.Q / (epsilon * A_w * rho_l * h_fg)

# Eq. (2-22): F_l = mu_l / (K * A_w * lambda * rho_l)
Fl = mu / (K * A_w * h_fg * rho_l)

# ------------------------------------------------------------------------
# Liquid pressure gradient, Eq. (2-21): dP_l/dx = -F_l*Q +/- rho_l*g*sin(psi)
# ------------------------------------------------------------------------
# SIGN CONVENTION (per Eq. 2-17: the gravity term is + or - depending on
# whether the liquid flows WITH or AGAINST gravity):
#
#   Geometry: evaporator at the BOTTOM, condenser at the TOP.
#   Chi's convention for a conventional heat pipe in heat-pipe mode
#   (Sec. 2-6) places x = 0 at the condenser end and x = L_t at the
#   evaporator end, with liquid flowing in the direction of increasing x
#   (condenser -> evaporator). Here that direction is DOWNWARD, i.e. the
#   SAME direction as gravity. Gravity therefore assists liquid return
#   and REDUCES the pumping pressure the wick must supply, which shows
#   up as a term with the same sign as -F_l*Q (both drive/assist flow in
#   the direction of integration), hence the minus sign below.
#
#   With psi = pi/2 (vertical) this gravity term is now nonzero and
#   noticeably changes the pressure gradient relative to the psi = 0
#   case used previously.
dP_l_dx = -(Fl * hp.Q) - (rho_l * hp.g * np.sin(hp.psi))

assert np.shape(dP_l_dx) == np.shape(T), \
    f"dP_l_dx came out as shape {np.shape(dP_l_dx)}, expected {np.shape(T)}"

# ------------------------------------------------------------------------
# Display results
# ------------------------------------------------------------------------
print("Liquid-pressure calculation completed.")
print("-----------------------------------------")
print(f"porosity                       : {epsilon:.4f} ")
print(f"permeability                   : {K:.4e} m^2")
print(f"Mesh Thickness                 : {hp.T_w:.4e} m")
print(f"Vapor Diameter                 : {dv:.4e} m")
print(f"Mesh Cross-Sectional Area      : {A_w:.4e} m^2")
print(f"Liquid Velocity range          : {np.min(Vl):.4f} to {np.max(Vl):.4f} m/s")
print(f"Frictional Coefficient         : {np.min(Fl):.4f} to {np.max(Fl):.4f} (N/m2)/(W-m)")
print(f"Last liquid pressure gradient  : {dP_l_dx[-1]:.4e} Pa/m")

# ------------------------------------------------------------------------
# Plot — SIMPLIFIED VIEW ONLY
# ------------------------------------------------------------------------
# NOTE: This plot assumes a spatially UNIFORM pressure gradient (i.e. Q
# constant along x) at each plotted temperature, so each curve is a
# straight line dP_l(x) = |dP_l/dx|_T * x. This is NOT the piecewise
# evaporator/adiabatic/condenser heat-flux distribution of Eq. (2-64); it
# is a simplified constant-Q visualization for a quick look at how the
# gradient scales with temperature. The rigorous zone-by-zone integration
# (with the correct effective-length method, Eq. 2-65) belongs in the
# capillary-limit kernel.
x = np.linspace(0, hp.Lt, 200)
T_interval = 200
plt.figure(figsize=(9, 6))
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    gradient = abs(dP_l_dx[i])
    Delta_P_l_x = gradient * x
    plt.plot(x, Delta_P_l_x, linewidth=2, label=f"{T[i]:.0f} K")
plt.xlabel("Axial Length, x [m]")
plt.ylabel("Liquid Pressure Drop, ΔP_L [Pa] (simplified, constant-Q)")
plt.title("Liquid Pressure Drop Along Heat Pipe (uniform-gradient approximation)")
plt.grid(True)
plt.legend(title="Temperature")
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# VAPOR PRESSURE DROP CALCULATION
# ============================================================================
# Chi Eqs. (2-33) through (2-45), Table 2-3
# per Chi's derivation preceding Eq. 2-31).

# Force T to the full temperature array — don't rely on it being set
# correctly by an earlier cell in this kernel session.
T = hp.temperature_range()

# Monatomic lithium vapor — specific heat ratio.
# NOTE: uses the T-dependent module function rather than a hardcoded 1.67.
# Over this design's 1000-1600 K range this returns exactly 5/3 (pure
# monatomic ideal-gas regime, since T < 1620.12 K everywhere here), so the
# numeric result is unchanged from before — this fix is about not silently
# diverging from li.py if T_max is ever raised above 1620.12 K in the future.
gamma_v = li.vapor_specific_heat_ratio(T)

# Vapor-phase properties over temperature range
rho_v = li.vapor_density(T)
mu_v = li.vapor_dynamic_viscosity(T)
lam = li.latent_heat_vaporization(T)      # same quantity as h_fg, called lambda here
R_v = li.vapor_specific_gas_constant()

# Vapor-core geometry (circular cross section)
r_h_v = dv / 2.0 #Hydraulic Radius of the heat pipe 
A_v = np.pi * (dv ** 2) / 4.0 # Area of the vapor core cross-sectional area

# Vapor Reynolds number — Eq. (2-36)
Re_v = (2 * r_h_v * hp.Q) / (A_v * mu_v * lam)

# Vapor flow Mach number — Eq. (2-37)
M_v = hp.Q / (A_v * rho_v * lam * np.sqrt(gamma_v * R_v * T))
# ==============================================================================================================================

# ----------------------------------------------------------------------
# Select F_v and D_v per Table 2-3, elementwise across the temperature array
# ----------------------------------------------------------------------
laminar = Re_v <= 2300
low_mach = M_v <= 0.2

# Case 1: laminar, incompressible (Re<=2300, M<=0.2)
F_v_1 = 8 * mu_v / (r_h_v**2 * A_v * rho_v * lam)
D_v_1 = 1.33 / (A_v**2 * rho_v * lam**2)

# Case 2: laminar, compressible (Re<=2300, M>0.2)
F_v_2 = (8 * mu_v / (r_h_v**2 * A_v * rho_v * lam)) * (1 + (gamma_v - 1)/2 * M_v**2)**(-0.5)
D_v_2 = 1.33 / (A_v**2 * rho_v * lam**2)

# Case 3: turbulent, incompressible (Re>2300, M<=0.2)
F_v_3 = (0.019 * mu_v / (A_v * r_h_v**2 * rho_v * lam)) * (2 * r_h_v * hp.Q / (A_v * lam * mu_v))**0.75
D_v_3 = 1.0 / (A_v**2 * rho_v * lam**2)

# Case 4: turbulent, compressible (Re>2300, M>0.2)
F_v_4 = (0.019 * mu_v / (A_v * r_h_v**2 * rho_v * lam)) * (2 * r_h_v * hp.Q / (A_v * lam * mu_v))**0.75 \
        * (1 + (gamma_v - 1)/2 * M_v**2)**(-0.75)
D_v_4 = 1.0 / (A_v**2 * rho_v * lam**2)
# ==============================================================================================================================

conditions = [
    laminar & low_mach,
    laminar & ~low_mach,
    ~laminar & low_mach,
    ~laminar & ~low_mach,
]

F_v = np.select(conditions, [F_v_1, F_v_2, F_v_3, F_v_4])
D_v = np.select(conditions, [D_v_1, D_v_2, D_v_3, D_v_4])
# ==============================================================================================================================

# Vapor pressure gradient — Eq. (2-33)
# The dynamic-pressure term D_v*dQ^2/dx is dropped here. Strictly, this term
# is only exactly zero when integrated over the FULL pipe length with Q=0 at
# both ends (shown explicitly in Chi, p.53, just after Eq. 2-61); it is not
# generally zero pointwise. This kernel reports a local, constant-Q gradient
# as a diagnostic (consistent with the liquid-side kernel's simplification),
# not the rigorous piecewise-Q(x) integral — that belongs in the capillary
# limit kernel.
dP_v_dx = -(F_v * hp.Q)

assert np.shape(dP_v_dx) == np.shape(T), \
    f"dP_v_dx came out as shape {np.shape(dP_v_dx)}, expected {np.shape(T)}"
# ==============================================================================================================================

# Local plotting/reporting parameters — defined here rather than relying on
# whatever a previous kernel happened to leave in the notebook namespace.
T_interval = 200
x = np.linspace(0, hp.Lt, 200)

# Display results
print("Vapor-pressure calculation completed.")
print("-----------------------------------------")
print(f"Vapor hydraulic radius         : {r_h_v:.4e} m")
print(f"Vapor core area                : {A_v:.4e} m^2")
print()
print(f"{'T [K]':>8} {'Re_v':>10} {'M_v':>10} {'Regime':>22} {'dP_v/dx [Pa/m]':>16}")
print("-" * 72)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    regime = f"{'laminar' if laminar[i] else 'turbulent'}, {'low-Mach' if low_mach[i] else 'compressible'}"
    print(f"{T[i]:8.0f} {Re_v[i]:10.2f} {M_v[i]:10.4f} {regime:>22} {dP_v_dx[i]:16.4e}")
# ==============================================================================================================================

# Plot vapor pressure drop along length, same style as liquid-side plot
plt.figure(figsize=(9, 6))
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    gradient = abs(dP_v_dx[i])
    Delta_P_v_x = gradient * x
    plt.plot(x, Delta_P_v_x, linewidth=2, label=f"{T[i]:.0f} K")
plt.xlabel("Axial Length, x [m]")
plt.ylabel("Vapor Pressure Drop, ΔP_v [Pa]")
plt.title("Vapor Pressure Drop Along the length of the Heat Pipe")
plt.grid(True)
plt.legend(title="Temperature")
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# CAPILLARY LIMITATION (Qc) 
# ============================================================================
# Chi Eqs. (2-58a), (2-22), Table 2-3, (2-62)-(2-65)
T = hp.temperature_range()
# ----------------------------------------------------------------------
# Temperature-dependent properties (from lithium_properties.py)
# ----------------------------------------------------------------------
sigma = li.surface_tension(T)
rho_l_cap = li.liquid_density(T)
mu_l_cap = li.liquid_dynamic_viscosity(T)
h_fg_cap = li.latent_heat_vaporization(T)
mu_v_cap = li.vapor_dynamic_viscosity(T)
rho_v_cap = li.vapor_density(T)
lam_cap = h_fg_cap   # same physical quantity as h_fg

# =====================================================================================================================
# Maximum capillary pressure and effective capillary pressure
# ----------------------------------------------------------------------
P_cm = (2.0 * sigma) / r_c                                      # Eq. (2-4)   PART 1
Delta_P_perp = rho_l_cap * hp.g * dv * np.cos(hp.psi)           # perpendicular gravity term (2-58 & 2-58a) PART2
P_cm_e = P_cm - Delta_P_perp                                    # Eq. (2-58a)

# ----------------------------------------------------------------------
# Frictional coefficients
# ----------------------------------------------------------------------
F_l_cap = mu_l_cap / (K * A_w * rho_l_cap * h_fg_cap)            # Eq. (2-22)

# NOTE: F_v is NOT recomputed here. It is taken directly from the vapor-
# pressure-drop kernel (kernel 6), which already selects the correct
# regime-dependent form (Table 2-3, Cases 1-4) based on the actual Re_v and
# M_v at each temperature. Recomputing a Case-1-only (laminar,
# incompressible) F_v here would silently disagree with kernel 6 at every
# temperature where the flow is compressible and/or turbulent -- which, for
# this design, is a significant part of the low-temperature end of the
# range (see kernel 6's Re_v/M_v table).
F_v_cap = F_v

# ----------------------------------------------------------------------
# Effective length — Eq. (2-64)
# ----------------------------------------------------------------------
L_eff = 0.5 * hp.Lc + hp.La + 0.5 * hp.Le

# ----------------------------------------------------------------------
# Axial hydrostatic term — Eq. (2-62)
# ----------------------------------------------------------------------
# SIGN CONVENTION: Chi's Eq. (2-62) is derived for a conventional heat pipe
# with the evaporator level with or ABOVE the condenser (Fig. 2-10), i.e.
# liquid returns UPHILL against gravity. There, the axial gravity term is
# subtracted, since gravity opposes liquid return and reduces the capillary
# pressure available for circulation.
#
# This design is the OPPOSITE orientation: evaporator at the BOTTOM,
# condenser at the TOP, so liquid returns DOWNHILL, i.e. gravity ASSISTS
# liquid return. Gravity therefore adds to (rather than subtracts from) the
# pressure available to drive circulation, so the sign of this term must be
# flipped relative to the textbook's literal Eq. (2-62) to correctly model
# this gravity-assisted case. See kernel 5's dP_l_dx sign derivation for the
# same reasoning applied to the differential form.
axial_gravity_term = rho_l_cap * hp.g * hp.Lt * np.sin(hp.psi)           # PART 3

# =====================================================================================================================
# Maximum heat transport factor — Eq. (2-63) Capillary Heat Transport Factor
# ----------------------------------------------------------------------
QL_c_max = (P_cm_e + axial_gravity_term) / (F_l_cap + F_v_cap)

# =====================================================================================================================
# Capillary-limited maximum heat transport rate — Eq. (2-65)
# ----------------------------------------------------------------------
Qc_max = QL_c_max / L_eff
# =====================================================================================================================

assert np.shape(Qc_max) == np.shape(T), \
    f"Qc_max came out as shape {np.shape(Qc_max)}, expected {np.shape(T)}"

# ----------------------------------------------------------------------
# Display results
# ----------------------------------------------------------------------
print("Capillary limitation (Qc) calculation completed.")
print("-----------------------------------------")
print(f"Effective length L_eff          : {L_eff:.4f} m")
print(f"Capillary radius r_c            : {r_c:.4e} m")
print()
print(f"{'T [K]':>8} {'P_cm,e [Pa]':>14} {'F_l':>12} {'F_v':>12} {'Qc [W]':>14} "
      f"{'Margin (Qc-Q)':>16}")
print("-" * 82)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    margin = Qc_max[i] - hp.Q
    print(f"{T[i]:8.0f} {P_cm_e[i]:14.4e} {F_l_cap[i]:12.4e} {F_v_cap[i]:12.4e} "
          f"{Qc_max[i]:14.2f} {margin:16.2f}")

# ----------------------------------------------------------------------
# Plot Qc vs. temperature, with operating Q for reference
# ----------------------------------------------------------------------
plt.figure(figsize=(9, 6))
plt.plot(T, Qc_max, linewidth=2, color='tab:red', label='$Q_c$ (capillary limit)')
plt.axhline(hp.Q, color='k', linestyle='--', linewidth=2, label=f'Operating Q = {hp.Q} W')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Heat Transport Rate, Q [W]")
plt.title("Capillary Limit Qc max vs. Operating Temperature")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# EFFECTIVE THERMAL CONDUCTIVITY OF WICK STRUCTURE (Wrapped Screen)
# ============================================================================
# Chi Eq. (2-48) — wrapped-screen wick (Table 2-4)

T = hp.temperature_range()

k_l = li.liquid_thermal_conductivity(T)   
k_w = tu.thermal_conductivity(T)   # dynamic tungsten thermal conductivity, W/(m*K)

ke = (k_l * ((k_l + k_w) - (1 - epsilon) * (k_l - k_w))) / ((k_l + k_w) + (1 - epsilon) * (k_l - k_w))

print("Effective wick thermal conductivity (ke) calculation completed.")
print("-----------------------------------------")
print(f"{'T [K]':>8} {'k_l [W/m-K]':>14} {'k_w [W/m-K]':>14} {'ke [W/m-K]':>14}")
print("-" * 54)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    kw_val = k_w[i] if hasattr(k_w, '__len__') else k_w
    print(f"{T[i]:8.0f} {k_l[i]:14.4f} {kw_val:14.4f} {ke[i]:14.4f}")

# ----------------------------------------------------------------------
# Plot: k_l, k_w, and ke together vs. temperature
# ----------------------------------------------------------------------
plt.figure(figsize=(9, 6))
plt.plot(T, k_l, linewidth=2, color='tab:cyan', label='$k_l$ (liquid lithium)')
plt.plot(T, k_w, linewidth=2, color='tab:purple', label='$k_w$ (tungsten)')
plt.plot(T, ke, linewidth=2.5, color='black', linestyle='--', label='$k_e$ (effective, wick+liquid)')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Thermal Conductivity [W/m-K]")
plt.title("Effective Wick Thermal Conductivity vs. Its Constituents")
plt.grid(True, alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# HEAT PIPE THERMAL RESISTANCE NETWORK / TEMPERATURE CHARACTERISTICS
# ============================================================================
# Standard series resistance network: evaporator wall -> evaporator wick -> vapor -> condenser wick -> condenser wall.
# Chi Eqs. (2-81), (2-82). Uses ke (effective wick conductivity, kernel 8)
# and kp (wick/wall metal conductivity, tungsten).
T = hp.temperature_range()

# ----------------------------------------------------------------------
# Vapor-side properties needed for the Clausius-Clapeyron temperature drop
# ----------------------------------------------------------------------
rho_v_res = li.vapor_density(T)
lam_res = li.latent_heat_vaporization(T)

# ----------------------------------------------------------------------
# Geometry needed for Chi's Eq. (2-82) family — actual radius, not diameter
# ----------------------------------------------------------------------
r_o = hp.do / 2.0 #Pipe's Outer Radius
r_i = hp.di / 2.0 #Pipe's Inner Radius
t_p = (hp.do - hp.di) / 2.0     # pipe wall thickness
kp = tu.thermal_conductivity(T)   # dynamic tungsten conductivity, W/(m*K), array, wall/container material conductivity becuase both are made up of tungsten
Tot_T_w = hp.T_w * hp.n_wick_layers

# ----------------------------------------------------------------------
# Radial conduction resistances [K/W] — Eq. (2-82)
# ----------------------------------------------------------------------
R_p_e = (r_o * t_p) / (2 * hp.Le * kp) 
R_w_e = (r_o**2 * Tot_T_w) / (2 * hp.Le * r_i * ke)

# NOTE on F_v: reused directly from kernel 6 (F_v), NOT recomputed here.
# Kernel 6 already selects the correct regime-dependent form (Table 2-3,
# Cases 1-4) based on the actual Re_v and M_v at each temperature; a
# hardcoded Case-1-only recomputation would silently disagree with kernel 6
# at every temperature where the flow is compressible and/or turbulent, the
# same issue that was fixed in kernel 7's F_v_cap.
#
# NOTE on J: Eq. (2-82) as printed in Chi includes a "mechanical equivalent
# of heat" factor J, needed only because Chi's worked examples mix English
# thermal units (Btu) with English force units (lbf) -- J = 778 ft-lbf/Btu
# converts between them. This entire notebook (li.py, tu.py, hp.py) is
# self-consistently SI throughout (kg, m, s, K, Pa, W, J), where 1 J of
# heat and 1 J of mechanical work are the same unit by definition. In SI,
# J = 1 exactly, so it is omitted below rather than divided out. The
# previous version of this kernel divided by 788 (itself a typo for 778),
# which was never a valid unit conversion here in the first place — it
# shrank R_v by a factor of ~788x.
R_v = (np.pi * r_o**2 * F_v * (hp.Le/6 + hp.La + hp.Lc/6) * T) / (rho_v_res * lam_res)

R_w_c = (r_o**2 * Tot_T_w) / (2 * hp.Lc * r_i * ke)
R_p_c = (r_o * t_p) / (2 * hp.Lc * kp)

# ----------------------------------------------------------------------
# Total resistance and total temperature drop
# ----------------------------------------------------------------------
R_total = R_p_e + R_w_e + R_v + R_w_c + R_p_c
U_hp_p = 1 / R_total
Delta_T_total = hp.Q * R_total

assert np.shape(R_total) == np.shape(T), \
   f"R_total came out as shape {np.shape(R_total)}, expected {np.shape(T)}"

# ============================================================================
# PRINT: Pipe/wick geometry values for Eq. (2-82)
# ============================================================================
print(f"Outer radius, r_o                 : {r_o:.4f} m")
print(f"Inner radius, r_i                 : {r_i:.4f} m")
print(f"Pipe wall thickness, t_p          : {t_p:.7f} m")
print(f"Single-layer thickness            : {hp.T_w:.7f} m")
print(f"Total wick thickness, T_w         : {Tot_T_w:.7f} m")

print()
print(f"{'T [K]':>8} {'R_p,e [K/W]':>14} {'R_w,e [K/W]':>14} {'R_v [K/W]':>14} {'R_w,c [K/W]':>14} {'R_p,c [K/W]':>14}")
print("-" * 85)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    print(f"{T[i]:8.0f} {R_p_e[i]:14.4e} {R_w_e[i]:14.4e} {R_v[i]:14.4e} {R_w_c[i]:14.4e} {R_p_c[i]:14.4e}")
    
print()
print(f"{'T [K]':>8} {'R_Total [K/W]':>16} {'U_hp_p [W/m^2*K]':>10} {'Delta_T [K]':>12}")
print("-" * 56)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    print(f"{T[i]:8.0f} {R_total[i]:14.4e} {U_hp_p[i]:14.4e} {Delta_T_total[i]:12.5f}")

# ----------------------------------------------------------------------
# Plot 1: the five individual resistances vs. temperature (log scale --
# R_v spans several orders of magnitude relative to the others, especially
# at the low-temperature end where vapor density is very low)
# ----------------------------------------------------------------------
plt.plot(T, R_p_e, linewidth=2, label='$R_{p,e}$ (evaporator wall)')
plt.plot(T, R_w_e, linewidth=2, label='$R_{w,e}$ (evaporator wick)')
plt.plot(T, R_v, linewidth=2, label='$R_v$ (vapor space)')
plt.plot(T, R_w_c, linewidth=4, linestyle=':', alpha=0.6, label='$R_{w,c}$ (condenser wick)')
plt.plot(T, R_p_c, linewidth=4, linestyle=':', alpha=0.6, label='$R_{p,c}$ (condenser wall)')
plt.plot(T, R_total, linewidth=2.5, color='black', label='$R_{total}$')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Thermal Resistance [K/W]")
plt.yscale('log')
plt.title("Heat Pipe Resistance Network Components vs. Temperature\n"
          "($R_{p,e}$=$R_{p,c}$ and $R_{w,e}$=$R_{w,c}$ exactly, since $L_e = L_c$)")
plt.grid(True, which='both', alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# SONIC LIMITATION (Qs)
# ============================================================================
# Chi Eq. (3-10) — Levy's equation
#
#   Q_s,max = A_v * rho_v * lambda * [ gamma_v * R_v * T_v / (2*(gamma_v+1)) ]^(1/2)
#
# The sonic limit occurs when the vapor Mach number at the evaporator exit
# reaches unity (choked flow). Beyond this limit, increasing heat rejection
# at the condenser does not increase heat transport -- it only drives
# supersonic flow and large axial temperature gradients (Chi, Sec. 3-2).
# This is an independent, self-contained limit calculation -- it does not
# depend on F_l, F_v, K, or any wick-geometry variable, since sonic
# limitation is governed by the vapor core alone (Chi states explicitly
# that "the sonic limitation is not greatly influenced by any aspect of the
# wick structure except for the vapor core size").
# ============================================================================

T = hp.temperature_range()

# ----------------------------------------------------------------------
# Vapor properties over the operating-temperature range
# ----------------------------------------------------------------------
rho_v_sonic = li.vapor_density(T)
lam_sonic = li.latent_heat_vaporization(T)
gamma_v_sonic = li.vapor_specific_heat_ratio(T)
R_v_sonic = li.vapor_specific_gas_constant()

# ----------------------------------------------------------------------
# Sonic-limited heat transport rate — Eq. (3-10)
# ----------------------------------------------------------------------
Qs_max = A_v * rho_v_sonic * lam_sonic * np.sqrt(
    (gamma_v_sonic * R_v_sonic * T) / (2.0 * (gamma_v_sonic + 1.0))
)

assert np.shape(Qs_max) == np.shape(T), \
    f"Qs_max came out as shape {np.shape(Qs_max)}, expected {np.shape(T)}"

# ----------------------------------------------------------------------
# Display results
# ----------------------------------------------------------------------
print("Sonic limitation (Qs) calculation completed.")
print("-----------------------------------------")
print(f"Vapor core area A_v             : {A_v:.4e} m^2")
print()
print(f"{'T [K]':>8} {'rho_v [kg/m3]':>14} {'gamma_v':>10} {'Qs [W]':>16} {'Margin (Qs-Q)':>16}")
print("-" * 68)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    margin = Qs_max[i] - hp.Q
    print(f"{T[i]:8.0f} {rho_v_sonic[i]:14.4e} {gamma_v_sonic[i]:10.4f} {Qs_max[i]:16.2f} {margin:16.2f}")

# ----------------------------------------------------------------------
# Feasibility flag: temperatures where the design Q exceeds the sonic limit
# ----------------------------------------------------------------------
sonic_limited = hp.Q > Qs_max
if np.any(sonic_limited):
    T_sonic_floor = T[sonic_limited][-1]
    print()
    print(f"WARNING: Operating Q = {hp.Q} W EXCEEDS the sonic limit for "
          f"T <= {T_sonic_floor:.0f} K.")
    print("At these temperatures the required vapor Mach number would "
          "exceed 1 (choked/supersonic) -- 2000 W is not physically "
          "deliverable there, regardless of wick design.")
else:
    print()
    print(f"Operating Q = {hp.Q} W is below the sonic limit across the "
          f"entire temperature range {T[0]:.0f}-{T[-1]:.0f} K.")

# ----------------------------------------------------------------------
# Plot Qs vs. temperature, with operating Q for reference
# ----------------------------------------------------------------------
plt.figure(figsize=(9, 6))
plt.plot(T, Qs_max, linewidth=2, color='tab:blue', label='$Q_s$ (sonic limit)')
plt.axhline(hp.Q, color='k', linestyle='--', linewidth=2, label=f'Operating Q = {hp.Q} W')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Heat Transport Rate, Q [W]")
plt.yscale('log')
plt.title("Sonic Limit Qs max vs. Operating Temperature")
plt.grid(True, which='both')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# ENTRAINMENT LIMITATION (Qe)
# ============================================================================
# Chi Eq. (3-17) / Eq. (7-15)
#
#   Q_e,max = A_v * lambda * sqrt( sigma * rho_v / (2 * r_h,s) )
#
# where r_h,s is the hydraulic radius of the wick's surface pores. For the
# wrapped-screen wick specifically, Chi states explicitly (Sec. 7-5):
# "r_h,s is equal to half of the wire spacing" -- i.e. r_h,s = w/2, where w
# is the clear spacing between adjacent wires (NOT the same as the
# capillary radius r_c = (w+d)/2 from Table 2-1, which is a different
# quantity used for the capillary limit, not entrainment). r_h,s uses the
# wire spacing W_s already computed in the capillary-pressure kernel.
# ============================================================================

T = hp.temperature_range()

# ----------------------------------------------------------------------
# Hydraulic radius of the wick surface pores, wrapped-screen wick
# ----------------------------------------------------------------------
# W_s (clear wire spacing) was already computed in the max-capillary-
# pressure kernel: W_s = wire_pitch - dw = (1/N) - dw
r_h_s = W_s / 2.0

# ----------------------------------------------------------------------
# Fluid properties over the operating-temperature range
# ----------------------------------------------------------------------
sigma_ent = li.surface_tension(T)
rho_v_ent = li.vapor_density(T)
lam_ent = li.latent_heat_vaporization(T)

# ----------------------------------------------------------------------
# Entrainment-limited heat transport rate — Eq. (3-17)/(7-15)
# ----------------------------------------------------------------------
Qe_max = A_v * lam_ent * np.sqrt((sigma_ent * rho_v_ent) / (2.0 * r_h_s))

assert np.shape(Qe_max) == np.shape(T), \
    f"Qe_max came out as shape {np.shape(Qe_max)}, expected {np.shape(T)}"

# ----------------------------------------------------------------------
# Display results
# ----------------------------------------------------------------------
print("Entrainment limitation (Qe) calculation completed.")
print("-----------------------------------------")
print(f"Wire spacing W_s                : {W_s:.4e} m")
print(f"Hydraulic radius r_h,s (=W_s/2)  : {r_h_s:.4e} m")
print(f"Vapor core area A_v             : {A_v:.4e} m^2")
print()
print(f"{'T [K]':>8} {'sigma [N/m]':>14} {'rho_v [kg/m3]':>14} {'Qe [W]':>16} {'Margin (Qe-Q)':>16}")
print("-" * 74)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    margin = Qe_max[i] - hp.Q
    print(f"{T[i]:8.0f} {sigma_ent[i]:14.4e} {rho_v_ent[i]:14.4e} {Qe_max[i]:16.2f} {margin:16.2f}")

# ----------------------------------------------------------------------
# Feasibility flag: temperatures where the design Q exceeds the entrainment limit
# ----------------------------------------------------------------------
entrainment_limited = hp.Q > Qe_max
if np.any(entrainment_limited):
    print()
    print(f"WARNING: Operating Q = {hp.Q} W EXCEEDS the entrainment limit at "
          f"{np.sum(entrainment_limited)} of {len(T)} temperature points.")
else:
    print()
    print(f"Operating Q = {hp.Q} W is below the entrainment limit across the "
          f"entire temperature range {T[0]:.0f}-{T[-1]:.0f} K.")

# ----------------------------------------------------------------------
# Plot Qe vs. temperature, with operating Q for reference
# ----------------------------------------------------------------------
plt.figure(figsize=(9, 6))
plt.plot(T, Qe_max, linewidth=2, color='tab:green', label='$Q_e$ (entrainment limit)')
plt.axhline(hp.Q, color='k', linestyle='--', linewidth=2, label=f'Operating Q = {hp.Q} W')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Heat Transport Rate, Q [W]")
plt.title("Entrainment Limit Qe max vs. Operating Temperature")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# BOILING LIMITATION (Qb)
# ============================================================================
# Chi Eq. (3-24)
#
#   Q_b,max = (2*pi*Le*ke*Tv) / (lambda*rho_v*ln(ri/rv)) * (2*sigma/rn - Pc)
#
# The boiling limit is a limit on RADIAL heat flux at the evaporator (not
# axial, like the other three limits): above this heat transfer rate,
# vapor bubbles nucleate and grow within the wick at the evaporator,
# disrupting liquid circulation. It uses ke, the effective thermal
# conductivity of the liquid-saturated wick, already computed in kernel 8.
#
# rn (bubble nucleation radius) is taken as 1e-5 in. = 2.54e-7 m, Chi's
# recommended conservative value for a conventional (non-gas-loaded) heat
# pipe (Sec. 3-4, p.92).
#
# Pc (capillary pressure) is taken as P_cm_e, the effective maximum
# capillary pressure already computed in the capillary-limit kernel, since
# Chi's own worked examples show Pc is well approximated by P_cm for this
# purpose (Example 3-5: "the capillary pressure Pc is usually less than
# P_cm"). This keeps the full Eq. (3-24) rather than dropping the Pc term
# outright (the simplified Eq. 7-18 form, valid when P_cm << 2*sigma/rn).
# ============================================================================

T = hp.temperature_range()

# ----------------------------------------------------------------------
# Sanity check on borrowed variables — catch stale/scalar state from an
# earlier kernel run immediately, instead of an obscure IndexError deep
# in the print loop below.
# ----------------------------------------------------------------------
for _name, _val in [("ke", ke), ("P_cm_e", P_cm_e), ("dv", dv), ("r_i", r_i)]:
    if np.shape(np.asarray(_val)) not in (np.shape(T), ()):
        raise ValueError(
            f"{_name} has shape {np.shape(_val)}, which matches neither a "
            f"scalar nor the current T array shape {np.shape(T)}. Re-run "
            f"the kernel that defines {_name} (kernel 7 for P_cm_e, "
            f"kernel 8 for ke) with the current T before running this one."
        )
assert np.ndim(ke) > 0, \
    "ke is a scalar, not an array -- re-run kernel 8 (effective thermal " \
    "conductivity) before this kernel."
assert np.ndim(P_cm_e) > 0, \
    "P_cm_e is a scalar, not an array -- re-run kernel 7 (capillary " \
    "limitation) before this kernel."

# ----------------------------------------------------------------------
# Nucleation radius — conservative value for a conventional heat pipe
# ----------------------------------------------------------------------
r_n = 2.54e-7   # m (1e-5 in.), Chi Sec. 3-4

# ----------------------------------------------------------------------
# Geometry — reuses r_i (pipe inner radius) from the resistance-network
# kernel, and r_v = dv/2 from the wick/vapor-core geometry established
# earlier
# ----------------------------------------------------------------------
r_v_boil = dv / 2.0

# ----------------------------------------------------------------------
# Fluid / wick properties over the operating-temperature range
# ----------------------------------------------------------------------
sigma_boil = li.surface_tension(T)
rho_v_boil = li.vapor_density(T)
lam_boil = li.latent_heat_vaporization(T)
Tv_boil = T   # vapor temperature, same array

# ke (effective wick conductivity) from kernel 8; P_cm_e (effective
# capillary pressure) from the capillary-limit kernel -- both reused
# directly, not recomputed.

# ----------------------------------------------------------------------
# Boiling-limited heat transport rate — Eq. (3-24)
# ----------------------------------------------------------------------
Qb_max = (
    (2.0 * np.pi * hp.Le * ke * Tv_boil)
    / (lam_boil * rho_v_boil * np.log(r_i / r_v_boil))
) * ((2.0 * sigma_boil / r_n) - P_cm_e)

assert np.shape(Qb_max) == np.shape(T), \
    f"Qb_max came out as shape {np.shape(Qb_max)}, expected {np.shape(T)}"

# ----------------------------------------------------------------------
# Display results
# ----------------------------------------------------------------------
print("Boiling limitation (Qb) calculation completed.")
print("-----------------------------------------")
print(f"Nucleation radius r_n            : {r_n:.4e} m")
print(f"Pipe inner radius r_i            : {r_i:.4e} m")
print(f"Vapor core radius r_v            : {r_v_boil:.4e} m")
print()
print(f"{'T [K]':>8} {'ke [W/m-K]':>12} {'2*sigma/rn [Pa]':>16} {'P_cm,e [Pa]':>14} {'Qb [W]':>16} {'Margin (Qb-Q)':>16}")
print("-" * 92)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    margin = Qb_max[i] - hp.Q
    two_sigma_rn = 2.0 * sigma_boil[i] / r_n
    print(f"{T[i]:8.0f} {ke[i]:12.4f} {two_sigma_rn:16.4e} {P_cm_e[i]:14.4e} {Qb_max[i]:16.2f} {margin:16.2f}")

# ----------------------------------------------------------------------
# Feasibility flag: temperatures where the design Q exceeds the boiling limit
# ----------------------------------------------------------------------
boiling_limited = hp.Q > Qb_max
if np.any(boiling_limited):
    print()
    print(f"WARNING: Operating Q = {hp.Q} W EXCEEDS the boiling limit at "
          f"{np.sum(boiling_limited)} of {len(T)} temperature points.")
else:
    print()
    print(f"Operating Q = {hp.Q} W is below the boiling limit across the "
          f"entire temperature range {T[0]:.0f}-{T[-1]:.0f} K.")

# ----------------------------------------------------------------------
# Plot Qb vs. temperature, with operating Q for reference
# ----------------------------------------------------------------------
plt.figure(figsize=(9, 6))
plt.plot(T, Qb_max, linewidth=2, color='tab:gray', label='$Q_b$ (boiling limit)')
plt.axhline(hp.Q, color='k', linestyle='--', linewidth=2, label=f'Operating Q = {hp.Q} W')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Heat Transport Rate, Q [W]")
plt.title("Boiling Limit Qb max vs. Operating Temperature")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# VISCOUS LIMITATION (Qv)
# ============================================================================
# NOTE: This limit is NOT covered in S. W. Chi, "Heat Pipe Theory and
# Practice" (the primary textbook used throughout this notebook). Chi's
# book addresses the same low-temperature, low-vapor-pressure startup
# failure mode through the sonic and entrainment limits directly (see
# Chi, Sec. 4-4 and Figs. 4-3/4-4/4-5), rather than as a separate named
# limit. The viscous (or "vapor pressure") limit as implemented below is
# a standard, independently-sourced result from the wider heat-pipe
# literature:
#
#   Q_v,max = (A_v * r_v^2 * lambda * rho_0 * P_0) / (16 * mu_v * L_eff)
#
# Reference: M. Busse, "Pressure Drop in the Vapor Phase of Long Heat
# Pipes," Proc. Intersociety Energy Conversion Engineering Conf.,
# pp. 391-398, 1967 (analytical treatment cited as Busse (1973) in later
# surveys); as summarized in, e.g.:
#   - Zhu & Vafai (misc. heat-pipe survey texts), and
#   - "Mathematical model for heat transfer limitations of heat pipe,"
#     Applied Thermal Engineering (ScienceDirect, S0895717711003888),
#   - arXiv:2411.11518, Eq. (2).
#
# Physical basis: the viscous limit occurs when the available vapor
# pressure at the evaporator is comparable to the pressure drop needed to
# drive vapor flow to the condenser. Busse's model assumes an isothermal
# ideal-gas vapor with condenser-end pressure taken as zero -- the
# limiting case in which the ENTIRE evaporator vapor pressure P_0 is
# consumed by viscous drag alone, giving the maximum possible heat
# transport under viscous-dominated (very low vapor density) conditions.
# This is most relevant at the low-temperature end of a liquid-metal heat
# pipe's operating range, exactly where this design's other limits (sonic,
# capillary) have already been flagged as marginal.
# ============================================================================

T = hp.temperature_range()

# ----------------------------------------------------------------------
# Vapor properties at the (closed) evaporator end, over the operating-
# temperature range. Per the Busse model's convention, evaluated using the
# bulk/static vapor properties at each T -- the same convention already
# used for rho_0, T_0 in the sonic-limit kernel.
# ----------------------------------------------------------------------
rho_0_visc = li.vapor_density(T)
P_0_visc = li.vapor_pressure(T)
mu_v_visc = li.vapor_dynamic_viscosity(T)
lam_visc = li.latent_heat_vaporization(T)

# ----------------------------------------------------------------------
# Geometry
# ----------------------------------------------------------------------
r_v_visc = dv / 2.0
L_eff_visc = 0.5 * hp.Lc + hp.La + 0.5 * hp.Le   # same effective length as the capillary limit, Eq. (2-64)

# ----------------------------------------------------------------------
# Viscous-limited heat transport rate — Busse's equation
# ----------------------------------------------------------------------
Qv_max = (A_v * r_v_visc**2 * lam_visc * rho_0_visc * P_0_visc) / (16.0 * mu_v_visc * L_eff_visc)

assert np.shape(Qv_max) == np.shape(T), \
    f"Qv_max came out as shape {np.shape(Qv_max)}, expected {np.shape(T)}"

# ----------------------------------------------------------------------
# Display results
# ----------------------------------------------------------------------
print("Viscous limitation (Qv) calculation completed.")
print("-----------------------------------------")
print("NOTE: not part of Chi's textbook framework -- see Busse citation above.")
print(f"Vapor core radius r_v            : {r_v_visc:.4e} m")
print(f"Effective length L_eff           : {L_eff_visc:.4f} m")
print()
print(f"{'T [K]':>8} {'rho_0 [kg/m3]':>14} {'P_0 [Pa]':>14} {'mu_v [Pa-s]':>14} {'Qv [W]':>16} {'Margin (Qv-Q)':>16}")
print("-" * 92)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    margin = Qv_max[i] - hp.Q
    print(f"{T[i]:8.0f} {rho_0_visc[i]:14.4e} {P_0_visc[i]:14.4e} {mu_v_visc[i]:14.4e} {Qv_max[i]:16.4e} {margin:16.4e}")

# ----------------------------------------------------------------------
# Feasibility flag: temperatures where the design Q exceeds the viscous limit
# ----------------------------------------------------------------------
viscous_limited = hp.Q > Qv_max
if np.any(viscous_limited):
    T_visc_floor = T[viscous_limited][-1]
    print()
    print(f"WARNING: Operating Q = {hp.Q} W EXCEEDS the viscous limit for "
          f"T <= {T_visc_floor:.0f} K.")
else:
    print()
    print(f"Operating Q = {hp.Q} W is below the viscous limit across the "
          f"entire temperature range {T[0]:.0f}-{T[-1]:.0f} K.")

# ----------------------------------------------------------------------
# Plot Qv vs. temperature, with operating Q for reference
# ----------------------------------------------------------------------
plt.figure(figsize=(9, 6))
plt.plot(T, Qv_max, linewidth=2, color='tab:orange', label='$Q_v$ (viscous limit, Busse)')
plt.axhline(hp.Q, color='k', linestyle='--', linewidth=2, label=f'Operating Q = {hp.Q} W')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Heat Transport Rate, Q [W]")
plt.yscale('log')
plt.title("Viscous Limit Qv max vs. Operating Temperature")
plt.grid(True, which='both')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# GOVERNING HEAT PIPE LIMIT vs. TEMPERATURE
# ============================================================================
# Chi, Chapter 3, Sec. 3-1: "The type of limitation restricting the heat
# transport capability of a heat pipe is determined by which limitation has
# the lowest value at the temperature under consideration."
#
# This kernel does NOT recompute any physics. It only combines the five
# independently-computed limit arrays already produced by earlier kernels:
#   Qc_max  -- capillary limit   (kernel 7)          [Chi, Eq. 2-65]
#   Qs_max  -- sonic limit       (sonic-limit kernel) [Chi, Eq. 3-10]
#   Qe_max  -- entrainment limit (entrainment kernel) [Chi, Eq. 3-17]
#   Qb_max  -- boiling limit     (boiling kernel)     [Chi, Eq. 3-24]
#   Qv_max  -- viscous limit     (viscous kernel)     [Busse, NOT in Chi]
# ============================================================================

T = hp.temperature_range()

# ----------------------------------------------------------------------
# Sanity check on borrowed variables — catch stale/scalar state from an
# earlier kernel run immediately, instead of an obscure error downstream.
# ----------------------------------------------------------------------
for _name, _val in [("Qc_max", Qc_max), ("Qs_max", Qs_max), ("Qe_max", Qe_max),
                     ("Qb_max", Qb_max), ("Qv_max", Qv_max)]:
    if np.ndim(_val) == 0 or np.shape(_val) != np.shape(T):
        raise ValueError(
            f"{_name} has shape {np.shape(_val)}, expected {np.shape(T)}. "
            f"Re-run the kernel that computes {_name} with the current T "
            f"before running this kernel."
        )

# ----------------------------------------------------------------------
# Stack the five limits and find the governing (minimum) one at each T
# ----------------------------------------------------------------------
labels = ['Capillary', 'Sonic', 'Entrainment', 'Boiling', 'Viscous']
limits = [Qc_max, Qs_max, Qe_max, Qb_max, Qv_max]
colors = np.array([
    [1.0, 0.0, 0.0],   # Capillary   - red
    [0.0, 0.0, 1.0],   # Sonic       - blue
    [0.0, 0.6, 0.0],   # Entrainment - green
    [0.5, 0.5, 0.5],   # Boiling     - gray
    [1.0, 0.55, 0.0],  # Viscous     - orange
])

Q_all = np.vstack(limits).T          # shape (n_T, 5)
Qgov = np.min(Q_all, axis=1)         # governing (minimum) limit at each T
idx_min = np.argmin(Q_all, axis=1)   # which limit governs at each T

# ----------------------------------------------------------------------
# Overall feasibility check against the design heat load
# ----------------------------------------------------------------------
feasible = hp.Q <= Qgov
if np.all(feasible):
    print(f"Operating Q = {hp.Q} W is FEASIBLE across the entire "
          f"temperature range {T[0]:.0f}-{T[-1]:.0f} K.")
else:
    T_infeasible = T[~feasible]
    print(f"Operating Q = {hp.Q} W EXCEEDS the governing limit for "
          f"{len(T_infeasible)} of {len(T)} temperature points "
          f"({T_infeasible.min():.0f} K to {T_infeasible.max():.0f} K).")
print()
print("NOTE: 'Viscous' is not part of Chi's textbook framework (see the "
      "viscous-limit kernel for its separate Busse citation); the other "
      "four limits are all from Chi's book.")
print()

# ----------------------------------------------------------------------
# Table: which limit governs, and over what temperature ranges
# ----------------------------------------------------------------------
from IPython.display import Markdown, display

governing_ranges = []
start_idx = 0
for i in range(1, len(T)):
    if idx_min[i] != idx_min[i - 1]:
        limit_id = idx_min[i - 1]
        end_idx = i - 1
        governing_ranges.append([
            labels[limit_id], T[start_idx], T[end_idx],
            Qgov[start_idx], Qgov[end_idx]
        ])
        start_idx = i
limit_id = idx_min[-1]
governing_ranges.append([
    labels[limit_id], T[start_idx], T[-1],
    Qgov[start_idx], Qgov[-1]
])

table = (
    "| Governing Limit | Temperature Start (K) | Temperature End (K) "
    "| Q Start (W) | Q End (W) |\n"
    "|:---:|:---:|:---:|:---:|:---:|\n"
)
for name, T_start, T_end, Q_start, Q_end in governing_ranges:
    table += f"| {name} | {T_start:.0f} | {T_end:.0f} | {Q_start:.1f} | {Q_end:.1f} |\n"

display(Markdown(table))

# ----------------------------------------------------------------------
# Plot: all five limits, governing envelope highlighted, operating Q shown
# ----------------------------------------------------------------------
# Each governing segment is extended a few points past its natural
# boundary (clipped to the array bounds) purely for visual clarity at
# crossover points -- this does NOT change which limit is reported as
# governing in the table above, only how the plot draws the handoff.
overlap_pts = 3   # extend each segment by this many T-points on each side

plt.figure(figsize=(10, 7))

for k, (label, limit_arr, color) in enumerate(zip(labels, limits, colors)):
    plt.plot(T, limit_arr, linestyle='--', linewidth=1.2, color=color,
              alpha=0.5, label=f'{label} (all T)')

used_labels = set()
start_idx = 0
for i in range(1, len(T)):
    if idx_min[i] != idx_min[i - 1]:
        limit_id = idx_min[i - 1]
        end_idx = i - 1

        plot_start = max(0, start_idx - overlap_pts)
        plot_end = min(len(T) - 1, end_idx + overlap_pts)

        lbl = f'{labels[limit_id]} (governing)' if labels[limit_id] not in used_labels else None
        used_labels.add(labels[limit_id])
        plt.plot(T[plot_start:plot_end + 1], limits[limit_id][plot_start:plot_end + 1],
                  '-', color=colors[limit_id], linewidth=3.0, label=lbl)
        start_idx = i

limit_id = idx_min[-1]
plot_start = max(0, start_idx - overlap_pts)
lbl = f'{labels[limit_id]} (governing)' if labels[limit_id] not in used_labels else None
plt.plot(T[plot_start:], limits[limit_id][plot_start:],
          '-', color=colors[limit_id], linewidth=3.0, label=lbl)

plt.axhline(hp.Q, color='k', linestyle=':', linewidth=2, label=f'Operating Q = {hp.Q} W')
plt.xlabel("Temperature, T [K]", fontsize=12)
plt.ylabel("Heat Transport Limit, Q [W]", fontsize=12)
plt.yscale('log')
plt.title("Governing Heat Pipe Limit vs. Temperature (5 limits)", fontsize=14)
plt.grid(True, which='both', alpha=0.4)
plt.legend(loc='best', fontsize=9)
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# AXIAL HEAT CONDUCTION THROUGH THE PIPE WALL (METAL)
# ============================================================================
# NOTE: This is NOT part of Chi's heat-pipe model. Chi's book (Sec. 2-7,
# Eq. 2-73 through 2-82) treats the pipe wall's conduction only in the
# RADIAL direction (evaporator wall -> wick, wick -> condenser wall), as
# part of the series resistance network already built in kernel 9. It does
# not model a separate AXIAL path where heat conducts directly along the
# solid tungsten wall from evaporator to condenser, bypassing the wick/
# vapor mechanism entirely.
#
# This kernel adds that separate, independent calculation using ordinary
# 1-D Fourier conduction through the pipe wall's annular cross-section:
#
#   Q_axial,wall = kp * A_wall,cross * Delta_T_total / Lt
#
# where Delta_T_total (= T_p,e - T_p,c) is the total end-to-end temperature
# drop already computed by the heat-pipe mechanism itself, in kernel 9.
# This tells you how much heat conducts directly through the metal wall,
# IN PARALLEL WITH (not instead of) the ~hp.Q that flows through the wick/
# vapor heat-pipe mechanism -- i.e., whether the pipe wall itself provides
# a meaningful heat-transport "short circuit," or whether it is negligible
# compared to the wick/vapor path.
#
# VALIDITY CAVEAT AT LOW TEMPERATURE: Delta_T_total (kernel 9) is driven
# largely by R_v, the vapor-space resistance, which is built on the same
# vapor pressure-drop formulas (Table 2-3) used in kernel 6. At the low-
# temperature end of this design's range, kernel 6 shows the vapor Mach
# number M_v exceeds 1 (choked/supersonic) -- a regime those formulas were
# never meant to be evaluated in. Consequently, Delta_T_total (and this
# kernel's Q_axial,wall) rises sharply at low T not because the metal wall
# is doing anything unusual, but because R_v itself is being computed from
# a vapor pressure-drop model outside its valid physical range. This is
# the SAME underlying issue already flagged by the sonic-limit kernel
# (Qs_max) and the viscous-limit kernel (Qv_max): at these temperatures,
# hp.Q is very likely NOT physically deliverable at all (see those
# kernels' feasibility warnings), so Delta_T_total and Q_axial,wall here
# should be read as "this operating point is not achievable," not as a
# literal prediction of pipe behavior.
# ============================================================================

T = hp.temperature_range()

# ----------------------------------------------------------------------
# Sanity check on borrowed variables from kernel 9
# ----------------------------------------------------------------------
for _name, _val in [("kp", kp), ("r_o", r_o), ("r_i", r_i), ("Delta_T_total", Delta_T_total)]:
    if np.ndim(_val) == 0 and _name not in ("r_o", "r_i"):
        raise ValueError(f"{_name} is a scalar -- re-run kernel 9 before this kernel.")

# ----------------------------------------------------------------------
# Pipe wall cross-sectional area (annulus), independent of temperature
# ----------------------------------------------------------------------
A_wall_cross = np.pi * (r_o**2 - r_i**2)

# ----------------------------------------------------------------------
# Axial conduction through the wall — simple Fourier conduction
# ----------------------------------------------------------------------
Q_axial_wall = kp * A_wall_cross * Delta_T_total / hp.Lt

# ----------------------------------------------------------------------
# Fraction of the design heat load this axial path represents
# ----------------------------------------------------------------------
axial_fraction_of_Q = Q_axial_wall / hp.Q

assert np.shape(Q_axial_wall) == np.shape(T), \
    f"Q_axial_wall came out as shape {np.shape(Q_axial_wall)}, expected {np.shape(T)}"

# ----------------------------------------------------------------------
# Cross-reference against sonic/viscous feasibility, if those kernels have
# been run in this session (optional -- only used for the printed note
# below, does not affect Q_axial_wall itself).
# ----------------------------------------------------------------------
_sonic_ok = 'Qs_max' in dir() and np.shape(Qs_max) == np.shape(T)
_viscous_ok = 'Qv_max' in dir() and np.shape(Qv_max) == np.shape(T)

# ----------------------------------------------------------------------
# Display results
# ----------------------------------------------------------------------
print("Axial wall conduction calculation completed.")
print("-----------------------------------------")
print("NOTE: not part of Chi's textbook model -- see derivation above.")
print(f"Pipe wall cross-sectional area A_wall  : {A_wall_cross:.4e} m^2")
print()
print(f"{'T [K]':>8} {'kp [W/m-K]':>12} {'Delta_T_total [K]':>18} {'Q_axial,wall [W]':>18} {'Fraction of Q':>16}")
print("-" * 78)
for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
    i = np.argmin(np.abs(T - T_plot))
    print(f"{T[i]:8.0f} {kp[i]:12.4f} {Delta_T_total[i]:18.5f} {Q_axial_wall[i]:18.6f} {axial_fraction_of_Q[i]:16.4%}")

print()
print(f"Axial wall conduction ranges from {Q_axial_wall.min():.4f} W to "
      f"{Q_axial_wall.max():.4f} W across the sweep, i.e. {axial_fraction_of_Q.min():.4%} "
      f"to {axial_fraction_of_Q.max():.4%} of the {hp.Q} W design heat load.")

if _sonic_ok or _viscous_ok:
    flagged = hp.Q > np.ones_like(T) * np.inf
    if _sonic_ok:
        flagged = flagged | (hp.Q > Qs_max)
    if _viscous_ok:
        flagged = flagged | (hp.Q > Qv_max)
    if np.any(flagged):
        T_flag = T[flagged]
        print()
        print(f"NOTE: For T <= {T_flag.max():.0f} K, hp.Q = {hp.Q} W exceeds the sonic "
              f"and/or viscous limit at that temperature (see those kernels). "
              f"The large Delta_T_total / Q_axial,wall values in that range "
              f"reflect an operating point that is not physically achievable, "
              f"not a real prediction of wall behavior.")
else:
    print()
    print("NOTE: Run the sonic-limit and/or viscous-limit kernels in this "
          "session to cross-check whether the low-temperature rise seen "
          "below corresponds to a physically infeasible operating point.")

# ----------------------------------------------------------------------
# Plot: axial wall conduction and its fraction of Q, vs. temperature
# ----------------------------------------------------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

ax1.plot(T, Q_axial_wall, linewidth=2, color='tab:brown')
ax1.set_xlabel("Temperature, T [K]")
ax1.set_ylabel("Axial Wall Conduction, Q_axial,wall [W]")
ax1.set_title("Axial Heat Conduction Through Pipe Wall")
ax1.set_yscale('log')
ax1.grid(True, which='both', alpha=0.4)

ax2.plot(T, axial_fraction_of_Q * 100.0, linewidth=2, color='tab:brown')
ax2.set_xlabel("Temperature, T [K]")
ax2.set_ylabel("Fraction of Operating Q [%]")
ax2.set_title(f"Axial Conduction as % of Q = {hp.Q} W")
ax2.set_yscale('log')
ax2.grid(True, which='both', alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# DESIGN VARIATION: NUMBER OF WICK LAYERS (n_wick_layers)
# ============================================================================
# Sweeps n_wick_layers over a set of candidate values, recomputing the
# geometry that depends on it (dv, A_v, r_h_v, A_w, Tot_T_w) and then all
# five heat-transport limits (Qc_max, Qs_max, Qe_max, Qb_max, Qv_max) at
# each candidate. epsilon, K, r_c, and W_s do NOT depend on n_wick_layers
# (they depend only on N, S, dw -- the mesh itself, not the number of
# layers), so those are computed once, outside the sweep, exactly as in
# kernels 4/5.
#
# This kernel recomputes the physics inline (does not call the individual
# limit kernels directly) so that all five limits can be evaluated at each
# of the several candidate geometries in one pass. The formulas used here
# are identical to, and should be kept in sync with, kernels 5-14.
# ============================================================================

import os

T = hp.temperature_range()

n_values = [3, 5, 7, 9, 12]

# ----------------------------------------------------------------------
# Output folder for saved figures
# ----------------------------------------------------------------------
# ----------------------------------------------------------------------
# Output folder for saved figures — your actual Windows Downloads folder
# ----------------------------------------------------------------------
output_dir = r"C:\Users\M.S.2\Downloads"
os.makedirs(output_dir, exist_ok=True)
# ----------------------------------------------------------------------
# Quantities independent of n_wick_layers (mesh properties only)
# ----------------------------------------------------------------------
wire_pitch = 1.0 / hp.N
W_s = wire_pitch - hp.dw
r_c = (W_s + hp.dw) / 2.0
r_h_s = W_s / 2.0

epsilon_sweep = 1.0 - (hp.S * np.pi * hp.N * hp.dw) / 4.0
K_sweep = (hp.dw ** 2) * (epsilon_sweep ** 3) / (122.0 * (1.0 - epsilon_sweep) ** 2)

# ----------------------------------------------------------------------
# Fluid properties (independent of geometry, computed once)
# ----------------------------------------------------------------------
sigma_T = li.surface_tension(T)
rho_l_T = li.liquid_density(T)
mu_l_T = li.liquid_dynamic_viscosity(T)
lam_T = li.latent_heat_vaporization(T)
mu_v_T = li.vapor_dynamic_viscosity(T)
rho_v_T = li.vapor_density(T)
P_v_T = li.vapor_pressure(T)
gamma_v_T = li.vapor_specific_heat_ratio(T)
R_v_val = li.vapor_specific_gas_constant()
kl_T = li.liquid_thermal_conductivity(T)
kw_T = tu.thermal_conductivity(T)

r_n = 2.54e-7   # boiling nucleation radius, Chi Sec. 3-4
L_eff = 0.5 * hp.Lc + hp.La + 0.5 * hp.Le
r_i = hp.di / 2.0

# ----------------------------------------------------------------------
# Storage for results
# ----------------------------------------------------------------------
results = {}   # results[n] = dict of limit arrays

for n in n_values:
    # Geometry that depends on n_wick_layers
    dv_n = hp.di - 2.0 * (n * hp.T_w)
    A_v_n = np.pi * (dv_n ** 2) / 4.0
    r_h_v_n = dv_n / 2.0
    A_w_n = np.pi * ((hp.di ** 2 - dv_n ** 2) / 4.0)
    r_v_n = dv_n / 2.0

    # Frictional coefficients
    F_l_n = mu_l_T / (K_sweep * A_w_n * rho_l_T * lam_T)

    Re_v_n = (2 * r_h_v_n * hp.Q) / (A_v_n * mu_v_T * lam_T)
    M_v_n = hp.Q / (A_v_n * rho_v_T * lam_T * np.sqrt(gamma_v_T * R_v_val * T))
    laminar_n = Re_v_n <= 2300
    low_mach_n = M_v_n <= 0.2

    F_v_1 = 8 * mu_v_T / (r_h_v_n**2 * A_v_n * rho_v_T * lam_T)
    F_v_2 = F_v_1 * (1 + (gamma_v_T - 1)/2 * M_v_n**2)**(-0.5)
    F_v_3 = (0.019 * mu_v_T / (A_v_n * r_h_v_n**2 * rho_v_T * lam_T)) * (2 * r_h_v_n * hp.Q / (A_v_n * lam_T * mu_v_T))**0.75
    F_v_4 = F_v_3 * (1 + (gamma_v_T - 1)/2 * M_v_n**2)**(-0.75)
    conditions_n = [laminar_n & low_mach_n, laminar_n & ~low_mach_n, ~laminar_n & low_mach_n, ~laminar_n & ~low_mach_n]
    F_v_n = np.select(conditions_n, [F_v_1, F_v_2, F_v_3, F_v_4])

    # Capillary limit
    P_cm_n = (2.0 * sigma_T) / r_c
    Delta_P_perp_n = rho_l_T * hp.g * dv_n * np.cos(hp.psi)
    P_cm_e_n = P_cm_n - Delta_P_perp_n
    axial_gravity_n = rho_l_T * hp.g * hp.Lt * np.sin(hp.psi)
    QL_c_max_n = (P_cm_e_n + axial_gravity_n) / (F_l_n + F_v_n)
    Qc_max_n = QL_c_max_n / L_eff

    # Sonic limit
    Qs_max_n = A_v_n * rho_v_T * lam_T * np.sqrt((gamma_v_T * R_v_val * T) / (2.0 * (gamma_v_T + 1.0)))

    # Entrainment limit
    Qe_max_n = A_v_n * lam_T * np.sqrt((sigma_T * rho_v_T) / (2.0 * r_h_s))

    # Effective wick conductivity (does not depend on n, but recomputed here for clarity)
    ke_n = (kl_T * ((kl_T + kw_T) - (1 - epsilon_sweep) * (kl_T - kw_T))) / \
           ((kl_T + kw_T) + (1 - epsilon_sweep) * (kl_T - kw_T))

    # Boiling limit
    Qb_max_n = (
        (2.0 * np.pi * hp.Le * ke_n * T) / (lam_T * rho_v_T * np.log(r_i / r_v_n))
    ) * ((2.0 * sigma_T / r_n) - P_cm_e_n)

    # Viscous limit
    Qv_max_n = (A_v_n * r_v_n**2 * lam_T * rho_v_T * P_v_T) / (16.0 * mu_v_T * L_eff)

    results[n] = {
        'dv': dv_n, 'A_v': A_v_n,
        'Qc_max': Qc_max_n, 'Qs_max': Qs_max_n, 'Qe_max': Qe_max_n,
        'Qb_max': Qb_max_n, 'Qv_max': Qv_max_n,
    }

# ----------------------------------------------------------------------
# 5 plots: one per limit, all n_wick_layers values overlaid.
# Each figure is saved as a PNG into the "downloads" folder before it is
# shown, using a filename derived from the limit name.
# ----------------------------------------------------------------------
limit_names = ['Qc_max', 'Qs_max', 'Qe_max', 'Qb_max', 'Qv_max']
limit_titles = ['Capillary Limit', 'Sonic Limit', 'Entrainment Limit', 'Boiling Limit', 'Viscous Limit']
limit_filenames = ['capillary_limit', 'sonic_limit', 'entrainment_limit', 'boiling_limit', 'viscous_limit']
n_colors = plt.cm.viridis(np.linspace(0, 1, len(n_values)))

saved_files = []

for limit_name, limit_title, limit_fname in zip(limit_names, limit_titles, limit_filenames):
    fig = plt.figure(figsize=(9, 6))
    for n, color in zip(n_values, n_colors):
        plt.plot(T, results[n][limit_name], linewidth=2, color=color, label=f'n = {n} layers')
    plt.axhline(hp.Q, color='k', linestyle='--', linewidth=1.5, label=f'Operating Q = {hp.Q} W')
    plt.xlabel("Temperature, T [K]")
    plt.ylabel("Heat Transport Rate, Q [W]")
    plt.yscale('log')
    plt.title(f"{limit_title} vs. Temperature — Wick Layer Count Variation")
    plt.grid(True, which='both', alpha=0.4)
    plt.legend()
    plt.tight_layout()

    filepath = os.path.join(output_dir, f"wick_layer_sweep_{limit_fname}.png")
    fig.savefig(filepath, dpi=200)
    saved_files.append(filepath)

    plt.show()

print("Saved figures:")
for f in saved_files:
    print(f"  {f}")

# ----------------------------------------------------------------------
# Table: governing limit and its value at each temperature, for each n
# ----------------------------------------------------------------------
from IPython.display import Markdown, display

table = "| n (layers) | T [K] | Governing Limit | Q_gov [W] | Q_c | Q_s | Q_e | Q_b | Q_v |\n"
table += "|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|\n"

for n in n_values:
    for T_plot in np.arange(T[0], T[-1] + 1, T_interval):
        i = np.argmin(np.abs(T - T_plot))
        vals = [results[n][name][i] for name in limit_names]
        gov_idx = np.argmin(vals)
        gov_name = limit_titles[gov_idx].replace(" Limit", "")
        row_vals = " | ".join(f"{v:.2e}" for v in vals)
        table += (f"| {n} | {T[i]:.0f} | {gov_name} | {vals[gov_idx]:.2e} | {row_vals} |\n")

display(Markdown(table))
```

```python
# ============================================================================
# KERNEL 17 — FULL COMPARISON: PURE W vs W-Re, KERNELS 4 THROUGH 16
# ============================================================================
# Recomputes EVERY calculation from kernels 4-16 for both container/wick
# materials and plots each one as a W vs W-Re comparison.
#
# Self-contained: it only needs the modules imported in kernel 2
# (np, plt, li, tu, wre, hp). It does NOT read or overwrite any variable
# from kernels 3-16 — every name here ends in _cmp, _W or _WRe, so your
# earlier kernels and their results are untouched.
#
# Same formulas as your kernels (Chi + Busse), with two corrections:
#   (1) Kernel 5: liquid gravity term uses "+ rho_l*g*sin(psi)" (gravity
#       ASSISTS return with evaporator at the bottom), consistent with
#       kernel 7. Your kernel 5 currently has "-".
#   (2) Kernels 9 & 15: Chi's Eq. (2-82) terms are AREA-BASED [m^2*K/W],
#       so dT = Q*R_total/(pi*r_o^2). Your kernel 9 uses dT = Q*R_total,
#       which makes dT (and kernel 15's Q_axial) ~1000x too small.
# So for pure W, kernel 5, 9 and 15 values here will differ from yours
# until those kernels are fixed; all other kernels should match exactly.
#
# Which results actually depend on the metal:
#   Material-INDEPENDENT (W and W-Re curves lie on top of each other):
#     K4 P_cm, K5 dP_l, K6 dP_v/Re_v/M_v, K7 Qc, K10 Qs, K11 Qe, K13 Qv
#   Material-DEPENDENT (W and W-Re differ):
#     K8 k_e, K9 R network & dT, K12 Qb, K14 governing, K15 axial,
#     K16 sweep (through Qb)
# ============================================================================

# ----------------------------------------------------------------------
# USER OPTIONS
# ----------------------------------------------------------------------
WICK_IS_WRE_cmp  = True    # True : wall AND wick screen are W-Re
                           # False: only the wall is W-Re, wick stays pure W
EXTEND_WRE_K_cmp = False   # False: material-dependent W-Re results only where
                           #        W-Re k data exist (373.15-1353.15 K) -> n/a above
                           # True : above 1353.15 K use the ASSUMPTION
                           #        k_WRe(T) = k_W(T) * [k_WRe/k_W at 1353.15 K]
r_n_cmp      = 2.54e-7               # nucleation radius [m] (Chi Sec. 3-4), as kernel 12
T_step_cmp   = 100                   # temperature step for printed tables [K]
n_values_cmp = [3, 5, 7, 9, 12]      # wick-layer sweep, as kernel 16

T_cmp = hp.temperature_range()

# ======================================================================
# METAL THERMAL CONDUCTIVITY: pure W and W-Re  [W/(m*K)]
# ======================================================================
k_W = tu.thermal_conductivity(T_cmp)

T_k_lo_cmp, T_k_hi_cmp = 373.15, 1353.15                # W-Re k data range
in_data_WRe  = (T_cmp >= T_k_lo_cmp) & (T_cmp <= T_k_hi_cmp)
k_WRe = np.full(T_cmp.shape, np.nan)
k_WRe[in_data_WRe] = wre.thermal_conductivity(T_cmp[in_data_WRe])

extended_WRe = np.zeros(T_cmp.shape, dtype=bool)
if EXTEND_WRE_K_cmp:
    k_ratio_cmp = float(wre.thermal_conductivity(T_k_hi_cmp)
                        / tu.thermal_conductivity(T_k_hi_cmp))
    extended_WRe = T_cmp > T_k_hi_cmp
    k_WRe[extended_WRe] = k_ratio_cmp * k_W[extended_WRe]

k_wall_WRe_cmp = k_WRe
k_wick_WRe_cmp = k_WRe if WICK_IS_WRE_cmp else k_W

# ======================================================================
# LITHIUM PROPERTIES (same for both materials)
# ======================================================================
Li_cmp = dict(
    sigma = li.surface_tension(T_cmp),
    rho_l = li.liquid_density(T_cmp),
    mu_l  = li.liquid_dynamic_viscosity(T_cmp),
    lam   = li.latent_heat_vaporization(T_cmp),
    k_l   = li.liquid_thermal_conductivity(T_cmp),
    rho_v = li.vapor_density(T_cmp),
    mu_v  = li.vapor_dynamic_viscosity(T_cmp),
    P_v   = li.vapor_pressure(T_cmp),
    gamma = li.vapor_specific_heat_ratio(T_cmp),
    R_v   = li.vapor_specific_gas_constant(),
)

lim_names_cmp = ["Capillary", "Sonic", "Entrainment", "Boiling", "Viscous"]

# ======================================================================
# FULL HEAT-PIPE MODEL (kernels 4-15) for one material / wick-layer count
# ======================================================================
def heat_pipe_model_cmp(k_wall, k_wick, n_layers):
    p, T, Q = Li_cmp, T_cmp, hp.Q
    o = {}

    # ---- Kernel 4: maximum capillary pressure, Eq. (2-4) --------------
    wire_pitch = 1.0 / hp.N
    W_s = wire_pitch - hp.dw
    r_c = (W_s + hp.dw) / 2.0
    o["r_c"], o["W_s"] = r_c, W_s
    o["P_cm"] = 2.0 * p["sigma"] / r_c

    # ---- Kernel 5: liquid pressure drop, Eqs. (2-19)-(2-28) -----------
    dv  = hp.di - 2.0 * (n_layers * hp.T_w)
    A_w = np.pi * (hp.di**2 - dv**2) / 4.0
    eps = 1.0 - (hp.S * np.pi * hp.N * hp.dw) / 4.0
    K   = hp.dw**2 * eps**3 / (122.0 * (1.0 - eps)**2)
    o.update(dv=dv, A_w=A_w, eps=eps, K=K)
    o["V_l"]    = Q / (eps * A_w * p["rho_l"] * p["lam"])
    o["F_l"]    = p["mu_l"] / (K * A_w * p["lam"] * p["rho_l"])
    o["dPl_dx"] = -(o["F_l"] * Q) + p["rho_l"] * hp.g * np.sin(hp.psi)   # corrected sign

    # ---- Kernel 6: vapor pressure drop, Eqs. (2-33)-(2-45), Table 2-3 -
    r_h_v = dv / 2.0
    A_v   = np.pi * dv**2 / 4.0
    o.update(r_h_v=r_h_v, A_v=A_v)
    Re_v = 2 * r_h_v * Q / (A_v * p["mu_v"] * p["lam"])
    M_v  = Q / (A_v * p["rho_v"] * p["lam"] * np.sqrt(p["gamma"] * p["R_v"] * T))
    lam_ = Re_v <= 2300
    lowM = M_v <= 0.2
    comp = 1 + (p["gamma"] - 1) / 2 * M_v**2
    Fv1 = 8 * p["mu_v"] / (r_h_v**2 * A_v * p["rho_v"] * p["lam"])
    Fv3 = (0.019 * p["mu_v"] / (A_v * r_h_v**2 * p["rho_v"] * p["lam"])) * Re_v**0.75
    F_v = np.select([lam_ & lowM, lam_ & ~lowM, ~lam_ & lowM, ~lam_ & ~lowM],
                    [Fv1, Fv1 * comp**-0.5, Fv3, Fv3 * comp**-0.75])
    o.update(Re_v=Re_v, M_v=M_v, F_v=F_v, laminar=lam_, low_mach=lowM)
    o["dPv_dx"] = -(F_v * Q)

    # ---- Kernel 7: capillary limit, Eqs. (2-58a), (2-62)-(2-65) -------
    L_eff = 0.5 * hp.Lc + hp.La + 0.5 * hp.Le
    o["L_eff"] = L_eff
    o["P_cm_e"] = o["P_cm"] - p["rho_l"] * hp.g * dv * np.cos(hp.psi)
    axial_g = p["rho_l"] * hp.g * hp.Lt * np.sin(hp.psi)     # gravity-assisted (+)
    o["Qc"] = (o["P_cm_e"] + axial_g) / (o["F_l"] + F_v) / L_eff

    # ---- Kernel 8: effective wick conductivity, Eq. (2-48) ------------
    kl = p["k_l"]
    o["k_wall"], o["k_wick"] = k_wall, k_wick
    o["ke"] = kl * ((kl + k_wick) - (1 - eps) * (kl - k_wick)) / \
                   ((kl + k_wick) + (1 - eps) * (kl - k_wick))

    # ---- Kernel 9: resistance network, Eqs. (2-81), (2-82) ------------
    r_o, r_i = hp.do / 2.0, hp.di / 2.0
    t_p = (hp.do - hp.di) / 2.0
    t_w = hp.T_w * n_layers
    A_hp = np.pi * r_o**2
    o.update(r_o=r_o, r_i=r_i, A_hp=A_hp)
    R = {
        "R_p,e": r_o * t_p / (2 * hp.Le * k_wall),
        "R_w,e": r_o**2 * t_w / (2 * hp.Le * r_i * o["ke"]),
        "R_v"  : np.pi * r_o**2 * F_v * (hp.Le/6 + hp.La + hp.Lc/6) * T
                 / (p["rho_v"] * p["lam"]),
        "R_w,c": r_o**2 * t_w / (2 * hp.Lc * r_i * o["ke"]),
        "R_p,c": r_o * t_p / (2 * hp.Lc * k_wall),
    }
    R["total"] = sum(R.values())                 # area-based [m^2*K/W]
    o["R"]  = R
    o["U"]  = 1.0 / R["total"]                   # [W/(m^2*K)]
    o["dT"] = Q * R["total"] / A_hp              # [K]  (corrected)

    # ---- Kernel 10: sonic limit, Eq. (3-10) ---------------------------
    o["Qs"] = A_v * p["rho_v"] * p["lam"] * np.sqrt(
        p["gamma"] * p["R_v"] * T / (2.0 * (p["gamma"] + 1.0)))

    # ---- Kernel 11: entrainment limit, Eq. (3-17) ---------------------
    o["Qe"] = A_v * p["lam"] * np.sqrt(p["sigma"] * p["rho_v"] / (2.0 * (W_s / 2.0)))

    # ---- Kernel 12: boiling limit, Eq. (3-24) -------------------------
    r_v = dv / 2.0
    o["Qb"] = (2.0 * np.pi * hp.Le * o["ke"] * T) / \
              (p["lam"] * p["rho_v"] * np.log(r_i / r_v)) * \
              (2.0 * p["sigma"] / r_n_cmp - o["P_cm_e"])

    # ---- Kernel 13: viscous limit (Busse) -----------------------------
    o["Qv"] = A_v * r_v**2 * p["lam"] * p["rho_v"] * p["P_v"] / (16.0 * p["mu_v"] * L_eff)

    # ---- Kernel 14: governing limit -----------------------------------
    lims = np.vstack([o["Qc"], o["Qs"], o["Qe"], o["Qb"], o["Qv"]])
    o["limits"] = lims
    o["Qgov"] = np.min(lims, axis=0)                              # NaN if no data
    o["gov"]  = np.array(["n/a" if np.any(np.isnan(c)) else lim_names_cmp[int(np.argmin(c))]
                          for c in lims.T])

    # ---- Kernel 15: axial conduction through the wall -----------------
    A_wall = np.pi * (r_o**2 - r_i**2)
    o["A_wall"] = A_wall
    o["Qax"] = k_wall * A_wall * o["dT"] / hp.Lt
    return o

# ======================================================================
# RUN THE MODEL: both materials at the design wick-layer count,
# plus the kernel-16 sweep for both materials
# ======================================================================
W_cmp   = heat_pipe_model_cmp(k_W,            k_W,            hp.n_wick_layers)
WRe_cmp = heat_pipe_model_cmp(k_wall_WRe_cmp, k_wick_WRe_cmp, hp.n_wick_layers)

sweep_W_cmp   = {n: heat_pipe_model_cmp(k_W, k_W, n) for n in n_values_cmp}
sweep_WRe_cmp = {n: heat_pipe_model_cmp(k_wall_WRe_cmp, k_wick_WRe_cmp, n)
                 for n in n_values_cmp}

# ======================================================================
# PRINTED TABLES — one per kernel, W and W-Re side by side
# ======================================================================
rows_cmp = [j for j in range(len(T_cmp))
            if np.isclose((T_cmp[j] - T_cmp[0]) % T_step_cmp, 0) or j == len(T_cmp) - 1]

def _cell_cmp(v, spec, w=12):
    if isinstance(v, str):
        return f"{v:>{w}}"
    return f"{'n/a':>{w}}" if np.isnan(v) else f"{format(v, spec):>{w}}"

def table_cmp(title, cols, note=""):
    """cols: list of (header, array_W, array_WRe, format_spec)."""
    print("\n" + "=" * 78)
    print(title + (f"   [{note}]" if note else ""))
    print("=" * 78)
    head = f"{'T [K]':>8}"
    for h, *_ in cols:
        head += f"{h + ' W':>14}{h + ' W-Re':>14}"
    print(head)
    print("-" * len(head))
    for j in rows_cmp:
        line = f"{T_cmp[j]:8.0f}"
        for _, aW, aR, spec in cols:
            line += _cell_cmp(aW[j], spec, 14) + _cell_cmp(aR[j], spec, 14)
        print(line + (" *" if extended_WRe[j] else ""))

print("FULL W vs W-Re COMPARISON (kernels 4-16) completed.")
print("-" * 78)
print(f"Wall material (case 2)   : W-Re")
print(f"Wick material (case 2)   : {'W-Re' if WICK_IS_WRE_cmp else 'pure W'}")
print(f"W-Re k above 1353 K      : "
      f"{'EXTENDED by scaling pure-W curve (ASSUMPTION, marked *)' if EXTEND_WRE_K_cmp else 'no data -> n/a'}")
print(f"Capillary radius r_c     : {W_cmp['r_c']:.4e} m")
print(f"Porosity / permeability  : {W_cmp['eps']:.4f} / {W_cmp['K']:.4e} m^2")
print(f"Vapor-core diameter dv   : {W_cmp['dv']:.4e} m")
print(f"Effective length L_eff   : {W_cmp['L_eff']:.4f} m")
print(f"Chi reference area pi*r_o^2 : {W_cmp['A_hp']:.4e} m^2")

table_cmp("K4  Maximum capillary pressure P_cm [Pa]",
          [("P_cm", W_cmp["P_cm"], WRe_cmp["P_cm"], ".4e")], "material-independent")
table_cmp("K5  Liquid pressure gradient dP_l/dx [Pa/m]",
          [("dPl/dx", W_cmp["dPl_dx"], WRe_cmp["dPl_dx"], ".4e")], "material-independent")
table_cmp("K6  Vapor flow: Re_v, M_v, dP_v/dx [Pa/m]",
          [("Re_v", W_cmp["Re_v"], WRe_cmp["Re_v"], ".2f"),
           ("M_v",  W_cmp["M_v"],  WRe_cmp["M_v"],  ".4f"),
           ("dPv/dx", W_cmp["dPv_dx"], WRe_cmp["dPv_dx"], ".3e")], "material-independent")
table_cmp("K7  Capillary limit Qc [W]",
          [("Qc", W_cmp["Qc"], WRe_cmp["Qc"], ".2f")], "material-independent")
table_cmp("K8  Wick-metal k and effective wick k_e [W/(m*K)]",
          [("k_wick", W_cmp["k_wick"], WRe_cmp["k_wick"], ".3f"),
           ("k_e",    W_cmp["ke"],     WRe_cmp["ke"],     ".3f")])
table_cmp("K9  Resistances (Chi area-based) [m^2*K/W]",
          [(key, W_cmp["R"][key], WRe_cmp["R"][key], ".3e")
           for key in ["R_p,e", "R_w,e", "R_v", "R_w,c", "R_p,c"]])
table_cmp("K9  Total resistance, U and temperature drop",
          [("R_tot", W_cmp["R"]["total"], WRe_cmp["R"]["total"], ".3e"),
           ("U",     W_cmp["U"],  WRe_cmp["U"],  ".3e"),
           ("dT[K]", W_cmp["dT"], WRe_cmp["dT"], ".4f")])
table_cmp("K10 Sonic limit Qs [W]",
          [("Qs", W_cmp["Qs"], WRe_cmp["Qs"], ".2f")], "material-independent")
table_cmp("K11 Entrainment limit Qe [W]",
          [("Qe", W_cmp["Qe"], WRe_cmp["Qe"], ".2f")], "material-independent")
table_cmp("K12 Boiling limit Qb [W]",
          [("Qb", W_cmp["Qb"], WRe_cmp["Qb"], ".4e")])
table_cmp("K13 Viscous limit Qv [W]",
          [("Qv", W_cmp["Qv"], WRe_cmp["Qv"], ".4e")], "material-independent")
table_cmp("K14 Governing limit",
          [("Qgov", W_cmp["Qgov"], WRe_cmp["Qgov"], ".3e"),
           ("Gov.", W_cmp["gov"],  WRe_cmp["gov"],  "")])
table_cmp("K15 Axial wall conduction [W] and fraction of Q [%]",
          [("Qax", W_cmp["Qax"], WRe_cmp["Qax"], ".4e"),
           ("%Q",  W_cmp["Qax"] / hp.Q * 100, WRe_cmp["Qax"] / hp.Q * 100, ".4f")])

# ---- K14 governing ranges, like your kernel 14 table ------------------
def gov_ranges_cmp(res):
    out, start = [], 0
    g = res["gov"]
    for j in range(1, len(T_cmp) + 1):
        if j == len(T_cmp) or g[j] != g[j - 1]:
            out.append((g[start], T_cmp[start], T_cmp[j - 1],
                        res["Qgov"][start], res["Qgov"][j - 1]))
            start = j
    return out

for lbl_cmp, res_cmp in [("pure W", W_cmp), ("W-Re", WRe_cmp)]:
    print(f"\nK14 Governing-limit ranges, {lbl_cmp}:")
    print(f"  {'Limit':<12} {'T start':>8} {'T end':>8} {'Q start [W]':>13} {'Q end [W]':>13}")
    for nm, t0, t1, q0, q1 in gov_ranges_cmp(res_cmp):
        print(f"  {nm:<12} {t0:8.0f} {t1:8.0f} {_cell_cmp(q0, '.1f', 13)} {_cell_cmp(q1, '.1f', 13)}")

# ---- K16 sweep table: governing limit for each n, both materials -------
print("\n" + "=" * 78)
print("K16 Wick-layer sweep: governing limit [W] (limit name)")
print("=" * 78)
print(f"{'n':>3} {'T [K]':>7} {'Qgov W':>12} {'Gov. W':>12} {'Qgov W-Re':>12} {'Gov. W-Re':>12}")
for n in n_values_cmp:
    for j in rows_cmp:
        a, b = sweep_W_cmp[n], sweep_WRe_cmp[n]
        print(f"{n:>3} {T_cmp[j]:7.0f} {_cell_cmp(a['Qgov'][j], '.3e')} {a['gov'][j]:>12} "
              f"{_cell_cmp(b['Qgov'][j], '.3e')} {b['gov'][j]:>12}"
              + (" *" if extended_WRe[j] else ""))

# ---- Summary of differences --------------------------------------------
ok_cmp = ~np.isnan(WRe_cmp["Qgov"])
if np.any(ok_cmp):
    print(f"\nLargest |W-Re / W - 1| over {T_cmp[ok_cmp][0]:.0f}-{T_cmp[ok_cmp][-1]:.0f} K:")
    for lbl_cmp, key_cmp in [("k_wall", "k_wall"), ("k_e", "ke"), ("dT", "dT"),
                             ("Qb", "Qb"), ("Qgov", "Qgov"), ("Qax", "Qax"),
                             ("P_cm", "P_cm"), ("Qc", "Qc"), ("Qs", "Qs"),
                             ("Qe", "Qe"), ("Qv", "Qv")]:
        d = np.nanmax(np.abs(WRe_cmp[key_cmp] / W_cmp[key_cmp] - 1)) * 100
        print(f"  {lbl_cmp:<7}: {d:8.3f} %")
print(f"\nFeasible points at Q = {hp.Q} W: pure W {np.sum(hp.Q <= W_cmp['Qgov'])}/{len(T_cmp)}, "
      f"W-Re {np.sum(hp.Q <= WRe_cmp['Qgov'])}/{np.sum(ok_cmp)} (points with W-Re data)")

# ======================================================================
# PLOTS — one independent figure per result; solid = W-Re, dashed = pure W
# ======================================================================
case_cmp = f"wall: W-Re, wick: {'W-Re' if WICK_IS_WRE_cmp else 'pure W'}"

def _shade_cmp():
    if EXTEND_WRE_K_cmp and np.any(extended_WRe):
        plt.axvspan(T_k_hi_cmp, T_cmp[-1], color='tab:orange', alpha=0.12,
                    label='W-Re k extrapolated (assumption)')

def plot_cmp(key, ylabel, title, color, log=False, absval=False, Qline=False,
             independent=False):
    yW, yR = W_cmp[key], WRe_cmp[key]
    if absval:
        yW, yR = np.abs(yW), np.abs(yR)
    plt.figure(figsize=(9, 6))
    plt.plot(T_cmp, yW, '--', color=color, lw=3,   alpha=0.9, label='pure W')
    plt.plot(T_cmp, yR, '-',  color=color, lw=1.8, alpha=0.9, label='W-Re')
    if Qline:
        plt.axhline(hp.Q, color='k', ls=':', lw=2, label=f'Operating Q = {hp.Q} W')
    if not independent:
        _shade_cmp()
    plt.xlabel("Temperature, T [K]")
    plt.ylabel(ylabel)
    if log:
        plt.yscale('log')
    sub = "identical for W and W-Re (no wall-metal dependence)" if independent else case_cmp
    plt.title(f"{title}: W vs W-Re\n({sub})")
    plt.grid(True, which='both', alpha=0.4)
    plt.legend()
    plt.tight_layout()
    plt.show()

# K4
plot_cmp("P_cm", "Maximum Capillary Pressure, P_cm [Pa]",
         "Maximum Capillary Pressure", 'tab:blue', independent=True)
# K5
plot_cmp("dPl_dx", "Liquid Pressure Gradient, dP_l/dx [Pa/m]",
         "Liquid Pressure Gradient", 'tab:cyan', independent=True)
plot_cmp("V_l", "Liquid Velocity, V_l [m/s]",
         "Liquid Velocity in Wick", 'tab:cyan', independent=True)
# K6
plot_cmp("dPv_dx", "|Vapor Pressure Gradient|, |dP_v/dx| [Pa/m]",
         "Vapor Pressure Gradient", 'tab:gray', log=True, absval=True, independent=True)
plot_cmp("Re_v", "Vapor Reynolds Number, Re_v [-]",
         "Vapor Reynolds Number", 'tab:gray', log=True, independent=True)
plot_cmp("M_v", "Vapor Mach Number, M_v [-]",
         "Vapor Mach Number", 'tab:gray', log=True, independent=True)
# K7
plot_cmp("Qc", "Heat Transport Rate, Q [W]",
         "Capillary Limit Qc", 'tab:red', Qline=True, independent=True)

# K8: metal k and k_e together
plt.figure(figsize=(9, 6))
plt.plot(T_cmp, k_W,   'k--', lw=2, label='$k$ pure W')
plt.plot(T_cmp, k_WRe, 'k-',  lw=2, label='$k$ W-Re')
plt.plot(T_cmp, W_cmp["ke"],   '--', color='tab:purple', lw=2, label='$k_e$ wick, pure W')
plt.plot(T_cmp, WRe_cmp["ke"], '-',  color='tab:purple', lw=2, label='$k_e$ wick, W-Re')
plt.plot(T_cmp, Li_cmp["k_l"], ':',  color='tab:cyan',   lw=2, label='$k_l$ liquid lithium')
_shade_cmp()
plt.xlabel("Temperature, T [K]")
plt.ylabel("Thermal Conductivity [W/(m·K)]")
plt.title(f"Metal and Effective Wick Conductivity: W vs W-Re\n({case_cmp})")
plt.grid(True, alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()

# K9: resistance components (K/W)
R_cols_cmp = {"R_p,e": 'tab:blue', "R_w,e": 'tab:green', "R_v": 'tab:gray',
              "R_w,c": 'tab:olive', "R_p,c": 'tab:brown'}
R_tex_cmp  = {"R_p,e": '$R_{p,e}$ (evap. wall)', "R_w,e": '$R_{w,e}$ (evap. wick)',
              "R_v": '$R_v$ (vapor)', "R_w,c": '$R_{w,c}$ (cond. wick)',
              "R_p,c": '$R_{p,c}$ (cond. wall)'}
A_hp_cmp = W_cmp["A_hp"]
plt.figure(figsize=(9, 6))
for key_cmp, c_cmp in R_cols_cmp.items():
    plt.plot(T_cmp, W_cmp["R"][key_cmp] / A_hp_cmp,   '--', color=c_cmp, lw=1.5)
    plt.plot(T_cmp, WRe_cmp["R"][key_cmp] / A_hp_cmp, '-',  color=c_cmp, lw=2,
             label=R_tex_cmp[key_cmp])
plt.plot(T_cmp, W_cmp["R"]["total"] / A_hp_cmp,   'k--', lw=2,   label='$R_{total}$, pure W')
plt.plot(T_cmp, WRe_cmp["R"]["total"] / A_hp_cmp, 'k-',  lw=2.5, label='$R_{total}$, W-Re')
_shade_cmp()
plt.xlabel("Temperature, T [K]")
plt.ylabel("Thermal Resistance [K/W]")
plt.yscale('log')
plt.title(f"Resistance Network (solid W-Re, dashed W)\n({case_cmp})")
plt.grid(True, which='both', alpha=0.4)
plt.legend(fontsize=9)
plt.tight_layout()
plt.show()

plot_cmp("dT", "Total Temperature Drop, ΔT [K]",
         f"Total Temperature Drop at Q = {hp.Q} W", 'tab:red', log=True)
# K10-K13
plot_cmp("Qs", "Heat Transport Rate, Q [W]", "Sonic Limit Qs",
         'tab:blue', log=True, Qline=True, independent=True)
plot_cmp("Qe", "Heat Transport Rate, Q [W]", "Entrainment Limit Qe",
         'tab:green', Qline=True, independent=True)
plot_cmp("Qb", "Heat Transport Rate, Q [W]", "Boiling Limit Qb",
         'tab:gray', log=True, Qline=True)
plot_cmp("Qv", "Heat Transport Rate, Q [W]", "Viscous Limit Qv",
         'tab:orange', log=True, Qline=True, independent=True)

# K14: all limits + governing envelope
lim_cols_cmp = ['tab:red', 'tab:blue', 'tab:green', 'tab:gray', 'tab:orange']
plt.figure(figsize=(10, 7))
for nm_cmp, c_cmp, qW_cmp, qR_cmp in zip(lim_names_cmp, lim_cols_cmp,
                                         W_cmp["limits"], WRe_cmp["limits"]):
    plt.plot(T_cmp, qW_cmp, '--', color=c_cmp, lw=1.5, alpha=0.7, label=f'{nm_cmp}, pure W')
    plt.plot(T_cmp, qR_cmp, '-',  color=c_cmp, lw=1.2, alpha=0.7, label=f'{nm_cmp}, W-Re')
plt.plot(T_cmp, W_cmp["Qgov"],   'k--', lw=3, alpha=0.6, label='Governing, pure W')
plt.plot(T_cmp, WRe_cmp["Qgov"], 'k-',  lw=3, alpha=0.6, label='Governing, W-Re')
plt.axhline(hp.Q, color='k', ls=':', lw=2, label=f'Operating Q = {hp.Q} W')
_shade_cmp()
plt.xlabel("Temperature, T [K]", fontsize=12)
plt.ylabel("Heat Transport Limit, Q [W]", fontsize=12)
plt.yscale('log')
plt.title(f"Governing Heat-Pipe Limit: W vs W-Re\n({case_cmp})", fontsize=13)
plt.grid(True, which='both', alpha=0.4)
plt.legend(fontsize=8, ncol=2)
plt.tight_layout()
plt.show()

# K15
plt.figure(figsize=(9, 6))
plt.plot(T_cmp, W_cmp["Qax"]   / hp.Q * 100, '--', color='tab:brown', lw=2, label='pure W')
plt.plot(T_cmp, WRe_cmp["Qax"] / hp.Q * 100, '-',  color='tab:brown', lw=2, label='W-Re')
_shade_cmp()
plt.xlabel("Temperature, T [K]")
plt.ylabel(f"Axial Wall Conduction [% of Q = {hp.Q} W]")
plt.yscale('log')
plt.title(f"Axial Heat Conduction Through the Pipe Wall: W vs W-Re\n({case_cmp})")
plt.grid(True, which='both', alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()

# K16: wick-layer sweep, one figure per limit + governing
n_cols_cmp = plt.cm.viridis(np.linspace(0, 1, len(n_values_cmp)))
for key_cmp, ttl_cmp in [("Qc", "Capillary Limit"), ("Qs", "Sonic Limit"),
                         ("Qe", "Entrainment Limit"), ("Qb", "Boiling Limit"),
                         ("Qv", "Viscous Limit"), ("Qgov", "Governing Limit")]:
    plt.figure(figsize=(9, 6))
    for n, c_cmp in zip(n_values_cmp, n_cols_cmp):
        plt.plot(T_cmp, sweep_W_cmp[n][key_cmp],   '--', color=c_cmp, lw=2.5, alpha=0.8)
        plt.plot(T_cmp, sweep_WRe_cmp[n][key_cmp], '-',  color=c_cmp, lw=1.5,
                 label=f'n = {n} layers')
    plt.plot([], [], 'k--', lw=2, label='pure W (dashed)')
    plt.plot([], [], 'k-',  lw=2, label='W-Re (solid)')
    plt.axhline(hp.Q, color='k', ls=':', lw=1.5, label=f'Operating Q = {hp.Q} W')
    indep = key_cmp in ("Qc", "Qs", "Qe", "Qv")
    if not indep:
        _shade_cmp()
    plt.xlabel("Temperature, T [K]")
    plt.ylabel("Heat Transport Rate, Q [W]")
    plt.yscale('log')
    plt.title(f"{ttl_cmp} vs. Wick Layer Count: W vs W-Re\n"
              f"({'identical for W and W-Re' if indep else case_cmp})")
    plt.grid(True, which='both', alpha=0.4)
    plt.legend(fontsize=9)
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================================
# KERNEL 18 — TUNGSTEN HEAT PIPE WITH SILVER WORKING FLUID (KERNELS 4-16)
# ============================================================================
# Recomputes every calculation from kernels 4-16 for:
#     wall + wick : pure tungsten   (tungsten_properties.py, imported as tu)
#     fluid       : silver          (silver_properties.py,   imported as ag)
#
# Self-contained: needs np, plt, tu, hp from kernel 2 and imports the silver
# module itself. It does NOT read or overwrite any variable from earlier
# kernels — every name here ends in _ag.
#
# Same formulas as kernel 17 (Chi + Busse), including the two corrections:
#   (1) liquid gravity term "+ rho_l*g*sin(psi)" (gravity-assisted return)
#   (2) dT = Q*R_total/(pi*r_o^2)  (Chi's Eq. 2-82 terms are area-based)
#
# TEMPERATURE RANGE: silver needs a much hotter pipe than lithium, so this
# kernel uses its own range (T_MIN_ag..T_MAX_ag) instead of
# hp.temperature_range(). Silver vapor pressure is ~1 kPa at 1800 K and
# ~100 kPa at its 2435 K boiling point.
#
# PROPERTY CAVEATS (see silver_properties.py header):
#   - mu_l, k_l, mu_v are PROVISIONAL estimates (one-time warnings)
#   - rho_l, sigma, mu_l are extrapolated above ~1570-1770 K
# ============================================================================
import warnings
import silver_properties as ag

# ----------------------------------------------------------------------
# USER OPTIONS
# ----------------------------------------------------------------------
T_MIN_ag, T_MAX_ag, T_INC_ag = 1800.0, 2400.0, 10.0   # silver operating range [K]
r_n_ag      = 2.54e-7               # nucleation radius [m] (Chi Sec. 3-4), as kernel 12
T_step_ag   = 100                   # temperature step for printed tables [K]
n_values_ag = [3, 5, 7, 9, 12]      # wick-layer sweep, as kernel 16

T_ag = np.arange(T_MIN_ag, T_MAX_ag + 0.5 * T_INC_ag, T_INC_ag)

# ======================================================================
# PROPERTIES: tungsten wall/wick + silver liquid/vapor
# ======================================================================
with warnings.catch_warnings():
    # Range/provisional warnings are summarised once below instead of
    # being repeated for every call.
    warnings.simplefilter("ignore")
    k_W_ag = tu.thermal_conductivity(T_ag)
    Ag_ag = dict(
        sigma = ag.surface_tension(T_ag),
        rho_l = ag.liquid_density(T_ag),
        mu_l  = ag.liquid_dynamic_viscosity(T_ag),
        lam   = ag.latent_heat_vaporization(T_ag),
        k_l   = ag.liquid_thermal_conductivity(T_ag),
        rho_v = ag.vapor_density(T_ag),
        mu_v  = ag.vapor_dynamic_viscosity(T_ag),
        P_v   = ag.vapor_pressure(T_ag),
        gamma = ag.vapor_specific_heat_ratio(T_ag),
        R_v   = ag.vapor_specific_gas_constant(),
    )

lim_names_ag = ["Capillary", "Sonic", "Entrainment", "Boiling", "Viscous"]

# ======================================================================
# FULL HEAT-PIPE MODEL (kernels 4-15) for a given wick-layer count
# ======================================================================
def heat_pipe_model_ag(n_layers):
    p, T, Q = Ag_ag, T_ag, hp.Q
    k_wall = k_wick = k_W_ag
    o = {}

    # ---- Kernel 4: maximum capillary pressure, Eq. (2-4) --------------
    wire_pitch = 1.0 / hp.N
    W_s = wire_pitch - hp.dw
    r_c = (W_s + hp.dw) / 2.0
    o["r_c"], o["W_s"] = r_c, W_s
    o["P_cm"] = 2.0 * p["sigma"] / r_c

    # ---- Kernel 5: liquid pressure drop, Eqs. (2-19)-(2-28) -----------
    dv  = hp.di - 2.0 * (n_layers * hp.T_w)
    A_w = np.pi * (hp.di**2 - dv**2) / 4.0
    eps = 1.0 - (hp.S * np.pi * hp.N * hp.dw) / 4.0
    K   = hp.dw**2 * eps**3 / (122.0 * (1.0 - eps)**2)
    o.update(dv=dv, A_w=A_w, eps=eps, K=K)
    o["V_l"]    = Q / (eps * A_w * p["rho_l"] * p["lam"])
    o["F_l"]    = p["mu_l"] / (K * A_w * p["lam"] * p["rho_l"])
    o["dPl_dx"] = -(o["F_l"] * Q) + p["rho_l"] * hp.g * np.sin(hp.psi)

    # ---- Kernel 6: vapor pressure drop, Eqs. (2-33)-(2-45), Table 2-3 -
    r_h_v = dv / 2.0
    A_v   = np.pi * dv**2 / 4.0
    o.update(r_h_v=r_h_v, A_v=A_v)
    Re_v = 2 * r_h_v * Q / (A_v * p["mu_v"] * p["lam"])
    M_v  = Q / (A_v * p["rho_v"] * p["lam"] * np.sqrt(p["gamma"] * p["R_v"] * T))
    lam_ = Re_v <= 2300
    lowM = M_v <= 0.2
    comp = 1 + (p["gamma"] - 1) / 2 * M_v**2
    Fv1 = 8 * p["mu_v"] / (r_h_v**2 * A_v * p["rho_v"] * p["lam"])
    Fv3 = (0.019 * p["mu_v"] / (A_v * r_h_v**2 * p["rho_v"] * p["lam"])) * Re_v**0.75
    F_v = np.select([lam_ & lowM, lam_ & ~lowM, ~lam_ & lowM, ~lam_ & ~lowM],
                    [Fv1, Fv1 * comp**-0.5, Fv3, Fv3 * comp**-0.75])
    o.update(Re_v=Re_v, M_v=M_v, F_v=F_v, laminar=lam_, low_mach=lowM)
    o["dPv_dx"] = -(F_v * Q)

    # ---- Kernel 7: capillary limit, Eqs. (2-58a), (2-62)-(2-65) -------
    L_eff = 0.5 * hp.Lc + hp.La + 0.5 * hp.Le
    o["L_eff"] = L_eff
    o["P_cm_e"] = o["P_cm"] - p["rho_l"] * hp.g * dv * np.cos(hp.psi)
    axial_g = p["rho_l"] * hp.g * hp.Lt * np.sin(hp.psi)     # gravity-assisted (+)
    o["Qc"] = (o["P_cm_e"] + axial_g) / (o["F_l"] + F_v) / L_eff

    # ---- Kernel 8: effective wick conductivity, Eq. (2-48) ------------
    kl = p["k_l"]
    o["ke"] = kl * ((kl + k_wick) - (1 - eps) * (kl - k_wick)) / \
                   ((kl + k_wick) + (1 - eps) * (kl - k_wick))

    # ---- Kernel 9: resistance network, Eqs. (2-81), (2-82) ------------
    r_o, r_i = hp.do / 2.0, hp.di / 2.0
    t_p = (hp.do - hp.di) / 2.0
    t_w = hp.T_w * n_layers
    A_hp = np.pi * r_o**2
    o.update(r_o=r_o, r_i=r_i, A_hp=A_hp)
    R = {
        "R_p,e": r_o * t_p / (2 * hp.Le * k_wall),
        "R_w,e": r_o**2 * t_w / (2 * hp.Le * r_i * o["ke"]),
        "R_v"  : np.pi * r_o**2 * F_v * (hp.Le/6 + hp.La + hp.Lc/6) * T
                 / (p["rho_v"] * p["lam"]),
        "R_w,c": r_o**2 * t_w / (2 * hp.Lc * r_i * o["ke"]),
        "R_p,c": r_o * t_p / (2 * hp.Lc * k_wall),
    }
    R["total"] = sum(R.values())                 # area-based [m^2*K/W]
    o["R"]  = R
    o["U"]  = 1.0 / R["total"]                   # [W/(m^2*K)]
    o["dT"] = Q * R["total"] / A_hp              # [K]

    # ---- Kernel 10: sonic limit, Eq. (3-10) ---------------------------
    o["Qs"] = A_v * p["rho_v"] * p["lam"] * np.sqrt(
        p["gamma"] * p["R_v"] * T / (2.0 * (p["gamma"] + 1.0)))

    # ---- Kernel 11: entrainment limit, Eq. (3-17) ---------------------
    o["Qe"] = A_v * p["lam"] * np.sqrt(p["sigma"] * p["rho_v"] / (2.0 * (W_s / 2.0)))

    # ---- Kernel 12: boiling limit, Eq. (3-24) -------------------------
    r_v = dv / 2.0
    o["Qb"] = (2.0 * np.pi * hp.Le * o["ke"] * T) / \
              (p["lam"] * p["rho_v"] * np.log(r_i / r_v)) * \
              (2.0 * p["sigma"] / r_n_ag - o["P_cm_e"])

    # ---- Kernel 13: viscous limit (Busse) -----------------------------
    o["Qv"] = A_v * r_v**2 * p["lam"] * p["rho_v"] * p["P_v"] / (16.0 * p["mu_v"] * L_eff)

    # ---- Kernel 14: governing limit -----------------------------------
    lims = np.vstack([o["Qc"], o["Qs"], o["Qe"], o["Qb"], o["Qv"]])
    o["limits"] = lims
    o["Qgov"] = np.min(lims, axis=0)
    o["gov"]  = np.array([lim_names_ag[int(np.argmin(c))] for c in lims.T])

    # ---- Kernel 15: axial conduction through the wall -----------------
    A_wall = np.pi * (r_o**2 - r_i**2)
    o["Qax"] = k_wall * A_wall * o["dT"] / hp.Lt
    return o

# ======================================================================
# RUN THE MODEL (design wick-layer count + kernel-16 sweep)
# ======================================================================
res_ag   = heat_pipe_model_ag(hp.n_wick_layers)
sweep_ag = {n: heat_pipe_model_ag(n) for n in n_values_ag}

# ======================================================================
# PRINTED TABLES
# ======================================================================
rows_ag = [j for j in range(len(T_ag))
           if np.isclose((T_ag[j] - T_ag[0]) % T_step_ag, 0) or j == len(T_ag) - 1]

def table_ag(title, cols):
    """cols: list of (header, array, format_spec)."""
    print("\n" + "=" * 78)
    print(title)
    print("=" * 78)
    head = f"{'T [K]':>8}" + "".join(f"{h:>15}" for h, *_ in cols)
    print(head)
    print("-" * len(head))
    for j in rows_ag:
        line = f"{T_ag[j]:8.0f}"
        for _, a, spec in cols:
            v = a[j]
            line += f"{v:>15}" if isinstance(v, str) else f"{format(v, spec):>15}"
        print(line)

print("TUNGSTEN / SILVER HEAT PIPE (kernels 4-16) completed.")
print("-" * 78)
print(f"Wall and wick            : pure tungsten")
print(f"Working fluid            : silver")
print(f"Temperature range        : {T_ag[0]:.0f} K to {T_ag[-1]:.0f} K, step {T_INC_ag:.0f} K")
print(f"Operating heat load Q    : {hp.Q} W")
print(f"Capillary radius r_c     : {res_ag['r_c']:.4e} m")
print(f"Porosity / permeability  : {res_ag['eps']:.4f} / {res_ag['K']:.4e} m^2")
print(f"Vapor-core diameter dv   : {res_ag['dv']:.4e} m")
print(f"Effective length L_eff   : {res_ag['L_eff']:.4f} m")
print(f"Chi reference area pi*r_o^2 : {res_ag['A_hp']:.4e} m^2")
print("NOTE: silver mu_l, k_l and mu_v are PROVISIONAL; rho_l, sigma and mu_l")
print("      are extrapolated above ~1570-1770 K (see silver_properties.py).")

table_ag("K4  Maximum capillary pressure",
         [("P_cm [Pa]", res_ag["P_cm"], ".4e")])
table_ag("K5  Liquid flow",
         [("dPl/dx [Pa/m]", res_ag["dPl_dx"], ".4e"),
          ("V_l [m/s]",     res_ag["V_l"],    ".4e"),
          ("F_l",           res_ag["F_l"],    ".4e")])
table_ag("K6  Vapor flow",
         [("Re_v",          res_ag["Re_v"],   ".2f"),
          ("M_v",           res_ag["M_v"],    ".4f"),
          ("F_v",           res_ag["F_v"],    ".4e"),
          ("dPv/dx [Pa/m]", res_ag["dPv_dx"], ".4e")])
table_ag("K7  Capillary limit",
         [("P_cm,e [Pa]", res_ag["P_cm_e"], ".4e"),
          ("Qc [W]",      res_ag["Qc"],     ".2f")])
table_ag("K8  Conductivities [W/(m*K)]",
         [("k_W",      k_W_ag,         ".3f"),
          ("k_l (Ag)", Ag_ag["k_l"],   ".3f"),
          ("k_e",      res_ag["ke"],   ".3f")])
table_ag("K9  Resistances (Chi area-based) [m^2*K/W]",
         [(key, res_ag["R"][key], ".3e") for key in ["R_p,e", "R_w,e", "R_v", "R_w,c", "R_p,c"]])
table_ag("K9  Total resistance, U and temperature drop",
         [("R_tot", res_ag["R"]["total"], ".3e"),
          ("U [W/m2K]", res_ag["U"],      ".3e"),
          ("dT [K]",    res_ag["dT"],     ".4f")])
table_ag("K10-K13  Operating limits [W]",
         [("Qs", res_ag["Qs"], ".3e"), ("Qe", res_ag["Qe"], ".3e"),
          ("Qb", res_ag["Qb"], ".3e"), ("Qv", res_ag["Qv"], ".3e")])
table_ag("K14 Governing limit",
         [("Qgov [W]", res_ag["Qgov"], ".3e"),
          ("Governing", res_ag["gov"], ""),
          ("Margin [W]", res_ag["Qgov"] - hp.Q, ".3e")])
table_ag("K15 Axial wall conduction",
         [("Qax [W]", res_ag["Qax"], ".4e"),
          ("% of Q",  res_ag["Qax"] / hp.Q * 100, ".4f")])

# ---- K14 governing ranges ----------------------------------------------
print("\nK14 Governing-limit ranges:")
print(f"  {'Limit':<12} {'T start':>8} {'T end':>8} {'Q start [W]':>13} {'Q end [W]':>13}")
start_ag = 0
for j in range(1, len(T_ag) + 1):
    if j == len(T_ag) or res_ag["gov"][j] != res_ag["gov"][j - 1]:
        print(f"  {res_ag['gov'][start_ag]:<12} {T_ag[start_ag]:8.0f} {T_ag[j-1]:8.0f} "
              f"{res_ag['Qgov'][start_ag]:13.1f} {res_ag['Qgov'][j-1]:13.1f}")
        start_ag = j

feasible_ag = hp.Q <= res_ag["Qgov"]
if np.all(feasible_ag):
    print(f"\nOperating Q = {hp.Q} W is FEASIBLE over {T_ag[0]:.0f}-{T_ag[-1]:.0f} K.")
else:
    print(f"\nOperating Q = {hp.Q} W EXCEEDS the governing limit at "
          f"{np.sum(~feasible_ag)} of {len(T_ag)} points "
          f"({T_ag[~feasible_ag].min():.0f}-{T_ag[~feasible_ag].max():.0f} K).")

# ---- K16 sweep table ----------------------------------------------------
print("\n" + "=" * 78)
print("K16 Wick-layer sweep: governing limit")
print("=" * 78)
print(f"{'n':>3} {'T [K]':>7} {'Qgov [W]':>12} {'Governing':>12}")
for n in n_values_ag:
    for j in rows_ag:
        print(f"{n:>3} {T_ag[j]:7.0f} {sweep_ag[n]['Qgov'][j]:12.3e} {sweep_ag[n]['gov'][j]:>12}")

# ======================================================================
# PLOTS — one independent figure per result
# ======================================================================
sub_ag = "tungsten wall/wick, silver working fluid"

def plot_ag(y, ylabel, title, color, log=False, absval=False, Qline=False):
    y = np.abs(y) if absval else y
    plt.figure(figsize=(9, 6))
    plt.plot(T_ag, y, '-', color=color, lw=2)
    if Qline:
        plt.axhline(hp.Q, color='k', ls='--', lw=2, label=f'Operating Q = {hp.Q} W')
        plt.legend()
    plt.xlabel("Temperature, T [K]")
    plt.ylabel(ylabel)
    if log:
        plt.yscale('log')
    plt.title(f"{title}\n({sub_ag})")
    plt.grid(True, which='both', alpha=0.4)
    plt.tight_layout()
    plt.show()

# K4
plot_ag(res_ag["P_cm"], "Maximum Capillary Pressure, P_cm [Pa]",
        "Maximum Capillary Pressure", 'tab:blue')
# K5
plot_ag(res_ag["dPl_dx"], "Liquid Pressure Gradient, dP_l/dx [Pa/m]",
        "Liquid Pressure Gradient", 'tab:cyan')
plot_ag(res_ag["V_l"], "Liquid Velocity, V_l [m/s]",
        "Liquid Velocity in Wick", 'tab:cyan')
# K6
plot_ag(res_ag["dPv_dx"], "|Vapor Pressure Gradient|, |dP_v/dx| [Pa/m]",
        "Vapor Pressure Gradient", 'tab:gray', log=True, absval=True)
plot_ag(res_ag["Re_v"], "Vapor Reynolds Number, Re_v [-]",
        "Vapor Reynolds Number", 'tab:gray', log=True)
plot_ag(res_ag["M_v"], "Vapor Mach Number, M_v [-]",
        "Vapor Mach Number", 'tab:gray', log=True)
# K7
plot_ag(res_ag["Qc"], "Heat Transport Rate, Q [W]",
        "Capillary Limit Qc", 'tab:red', Qline=True)

# K8
plt.figure(figsize=(9, 6))
plt.plot(T_ag, k_W_ag,       'k-',  lw=2, label='$k$ tungsten')
plt.plot(T_ag, Ag_ag["k_l"], ':',   color='tab:cyan',   lw=2, label='$k_l$ liquid silver (provisional)')
plt.plot(T_ag, res_ag["ke"], '--',  color='tab:purple', lw=2, label='$k_e$ wick (W + Ag)')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Thermal Conductivity [W/(m·K)]")
plt.title(f"Effective Wick Thermal Conductivity vs. Its Constituents\n({sub_ag})")
plt.grid(True, alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()

# K9: resistance components (K/W)
R_cols_ag = {"R_p,e": 'tab:blue', "R_w,e": 'tab:green', "R_v": 'tab:gray',
             "R_w,c": 'tab:olive', "R_p,c": 'tab:brown'}
R_tex_ag  = {"R_p,e": '$R_{p,e}$ (evap. wall)', "R_w,e": '$R_{w,e}$ (evap. wick)',
             "R_v": '$R_v$ (vapor)', "R_w,c": '$R_{w,c}$ (cond. wick)',
             "R_p,c": '$R_{p,c}$ (cond. wall)'}
plt.figure(figsize=(9, 6))
for key_ag, c_ag in R_cols_ag.items():
    ls_ag = ':' if key_ag in ("R_w,c", "R_p,c") else '-'
    plt.plot(T_ag, res_ag["R"][key_ag] / res_ag["A_hp"], ls_ag, color=c_ag,
             lw=3 if ls_ag == ':' else 2, label=R_tex_ag[key_ag])
plt.plot(T_ag, res_ag["R"]["total"] / res_ag["A_hp"], 'k-', lw=2.5, label='$R_{total}$')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Thermal Resistance [K/W]")
plt.yscale('log')
plt.title(f"Heat Pipe Resistance Network Components vs. Temperature\n({sub_ag})")
plt.grid(True, which='both', alpha=0.4)
plt.legend(fontsize=9)
plt.tight_layout()
plt.show()

plot_ag(res_ag["dT"], "Total Temperature Drop, ΔT [K]",
        f"Total Temperature Drop at Q = {hp.Q} W", 'tab:red', log=True)
# K10-K13
plot_ag(res_ag["Qs"], "Heat Transport Rate, Q [W]", "Sonic Limit Qs",
        'tab:blue', log=True, Qline=True)
plot_ag(res_ag["Qe"], "Heat Transport Rate, Q [W]", "Entrainment Limit Qe",
        'tab:green', Qline=True)
plot_ag(res_ag["Qb"], "Heat Transport Rate, Q [W]", "Boiling Limit Qb",
        'tab:gray', log=True, Qline=True)
plot_ag(res_ag["Qv"], "Heat Transport Rate, Q [W]", "Viscous Limit Qv",
        'tab:orange', log=True, Qline=True)

# K14: all limits + governing envelope
lim_cols_ag = ['tab:red', 'tab:blue', 'tab:green', 'tab:gray', 'tab:orange']
plt.figure(figsize=(10, 7))
for nm_ag, c_ag, q_ag in zip(lim_names_ag, lim_cols_ag, res_ag["limits"]):
    plt.plot(T_ag, q_ag, '--', color=c_ag, lw=1.5, alpha=0.7, label=nm_ag)
plt.plot(T_ag, res_ag["Qgov"], 'k-', lw=3, alpha=0.6, label='Governing limit')
plt.axhline(hp.Q, color='k', ls=':', lw=2, label=f'Operating Q = {hp.Q} W')
plt.xlabel("Temperature, T [K]", fontsize=12)
plt.ylabel("Heat Transport Limit, Q [W]", fontsize=12)
plt.yscale('log')
plt.title(f"Governing Heat Pipe Limit vs. Temperature (5 limits)\n({sub_ag})", fontsize=13)
plt.grid(True, which='both', alpha=0.4)
plt.legend(fontsize=9)
plt.tight_layout()
plt.show()

# K15
plot_ag(res_ag["Qax"] / hp.Q * 100, f"Axial Wall Conduction [% of Q = {hp.Q} W]",
        "Axial Heat Conduction Through the Pipe Wall", 'tab:brown', log=True)

# K16: wick-layer sweep, one figure per limit + governing
n_cols_ag = plt.cm.viridis(np.linspace(0, 1, len(n_values_ag)))
for key_ag, ttl_ag in [("Qc", "Capillary Limit"), ("Qs", "Sonic Limit"),
                       ("Qe", "Entrainment Limit"), ("Qb", "Boiling Limit"),
                       ("Qv", "Viscous Limit"), ("Qgov", "Governing Limit")]:
    plt.figure(figsize=(9, 6))
    for n, c_ag in zip(n_values_ag, n_cols_ag):
        plt.plot(T_ag, sweep_ag[n][key_ag], '-', color=c_ag, lw=2, label=f'n = {n} layers')
    plt.axhline(hp.Q, color='k', ls='--', lw=1.5, label=f'Operating Q = {hp.Q} W')
    plt.xlabel("Temperature, T [K]")
    plt.ylabel("Heat Transport Rate, Q [W]")
    plt.yscale('log')
    plt.title(f"{ttl_ag} vs. Wick Layer Count\n({sub_ag})")
    plt.grid(True, which='both', alpha=0.4)
    plt.legend(fontsize=9)
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================================
# KERNEL 19 — W-Re HEAT PIPE WITH SILVER WORKING FLUID (KERNELS 4-16)
# ============================================================================
# Recomputes every calculation from kernels 4-16 for:
#     wall        : W-Re      (tungsten_rhenium_properties.py, imported as wre)
#     wick        : W-Re or pure W (option WICK_IS_WRE_agr below)
#     fluid       : silver    (silver_properties.py, imported as ag)
#
# Self-contained: needs np, plt, tu, wre, hp from kernel 2 and imports the
# silver module itself. It does NOT read or overwrite any variable from
# earlier kernels (including kernel 18) — every name here ends in _agr.
#
# Same formulas as kernels 17/18 (Chi + Busse), including the two corrections:
#   (1) liquid gravity term "+ rho_l*g*sin(psi)" (gravity-assisted return)
#   (2) dT = Q*R_total/(pi*r_o^2)  (Chi's Eq. 2-82 terms are area-based)
#
# !! W-Re CONDUCTIVITY IS EXTRAPOLATED OVER THE WHOLE SILVER RANGE !!
# The W-Re thermal-conductivity data stop at 1353.15 K, but a silver pipe
# runs at ~1800-2400 K. This kernel therefore uses the ASSUMPTION
#     k_WRe(T) = k_W(T) * [k_WRe / k_W at 1353.15 K]
# i.e. the pure-W curve scaled to match the last W-Re data point.
# Every W-Re-dependent result (k_e, R_p, R_w, dT, Qb, Q_axial) rests on that
# assumption. Results that do not depend on the wall metal (P_cm, dP_l,
# dP_v, Qc, Qs, Qe, Qv) are identical to kernel 18.
#
# TEMPERATURE RANGE: silver needs a much hotter pipe than lithium, so this
# kernel uses its own range (T_MIN_agr..T_MAX_agr) instead of
# hp.temperature_range().
#
# PROPERTY CAVEATS (see silver_properties.py header):
#   - mu_l, k_l, mu_v are PROVISIONAL estimates
#   - rho_l, sigma, mu_l are extrapolated above ~1570-1770 K
# ============================================================================
import warnings
import silver_properties as ag

# ----------------------------------------------------------------------
# USER OPTIONS
# ----------------------------------------------------------------------
WICK_IS_WRE_agr = True              # True : wall AND wick screen are W-Re
                                    # False: only the wall is W-Re, wick stays pure W
T_MIN_agr, T_MAX_agr, T_INC_agr = 1800.0, 2400.0, 10.0   # silver operating range [K]
r_n_agr      = 2.54e-7               # nucleation radius [m] (Chi Sec. 3-4), as kernel 12
T_step_agr   = 100                   # temperature step for printed tables [K]
n_values_agr = [3, 5, 7, 9, 12]      # wick-layer sweep, as kernel 16

T_agr = np.arange(T_MIN_agr, T_MAX_agr + 0.5 * T_INC_agr, T_INC_agr)

# ======================================================================
# PROPERTIES: W-Re wall (+ wick) + silver liquid/vapor
# ======================================================================
with warnings.catch_warnings():
    # Range/provisional warnings are summarised once below instead of
    # being repeated for every call.
    warnings.simplefilter("ignore")
    k_W_agr = tu.thermal_conductivity(T_agr)
    T_k_hi_agr = 1353.15                                   # end of W-Re k data
    k_ratio_agr = float(wre.thermal_conductivity(T_k_hi_agr)
                        / tu.thermal_conductivity(T_k_hi_agr))
    k_WRe_agr = np.where(T_agr <= T_k_hi_agr,
                         wre.thermal_conductivity(np.clip(T_agr, 373.15, T_k_hi_agr)),
                         k_ratio_agr * k_W_agr)            # ASSUMPTION above 1353 K
    extrapolated_agr = T_agr > T_k_hi_agr
    Ag_agr = dict(
        sigma = ag.surface_tension(T_agr),
        rho_l = ag.liquid_density(T_agr),
        mu_l  = ag.liquid_dynamic_viscosity(T_agr),
        lam   = ag.latent_heat_vaporization(T_agr),
        k_l   = ag.liquid_thermal_conductivity(T_agr),
        rho_v = ag.vapor_density(T_agr),
        mu_v  = ag.vapor_dynamic_viscosity(T_agr),
        P_v   = ag.vapor_pressure(T_agr),
        gamma = ag.vapor_specific_heat_ratio(T_agr),
        R_v   = ag.vapor_specific_gas_constant(),
    )

lim_names_agr = ["Capillary", "Sonic", "Entrainment", "Boiling", "Viscous"]

# ======================================================================
# FULL HEAT-PIPE MODEL (kernels 4-15) for a given wick-layer count
# ======================================================================
def heat_pipe_model_agr(n_layers):
    p, T, Q = Ag_agr, T_agr, hp.Q
    k_wall = k_WRe_agr
    k_wick = k_WRe_agr if WICK_IS_WRE_agr else k_W_agr
    o = {}

    # ---- Kernel 4: maximum capillary pressure, Eq. (2-4) --------------
    wire_pitch = 1.0 / hp.N
    W_s = wire_pitch - hp.dw
    r_c = (W_s + hp.dw) / 2.0
    o["r_c"], o["W_s"] = r_c, W_s
    o["P_cm"] = 2.0 * p["sigma"] / r_c

    # ---- Kernel 5: liquid pressure drop, Eqs. (2-19)-(2-28) -----------
    dv  = hp.di - 2.0 * (n_layers * hp.T_w)
    A_w = np.pi * (hp.di**2 - dv**2) / 4.0
    eps = 1.0 - (hp.S * np.pi * hp.N * hp.dw) / 4.0
    K   = hp.dw**2 * eps**3 / (122.0 * (1.0 - eps)**2)
    o.update(dv=dv, A_w=A_w, eps=eps, K=K)
    o["V_l"]    = Q / (eps * A_w * p["rho_l"] * p["lam"])
    o["F_l"]    = p["mu_l"] / (K * A_w * p["lam"] * p["rho_l"])
    o["dPl_dx"] = -(o["F_l"] * Q) + p["rho_l"] * hp.g * np.sin(hp.psi)

    # ---- Kernel 6: vapor pressure drop, Eqs. (2-33)-(2-45), Table 2-3 -
    r_h_v = dv / 2.0
    A_v   = np.pi * dv**2 / 4.0
    o.update(r_h_v=r_h_v, A_v=A_v)
    Re_v = 2 * r_h_v * Q / (A_v * p["mu_v"] * p["lam"])
    M_v  = Q / (A_v * p["rho_v"] * p["lam"] * np.sqrt(p["gamma"] * p["R_v"] * T))
    lam_ = Re_v <= 2300
    lowM = M_v <= 0.2
    comp = 1 + (p["gamma"] - 1) / 2 * M_v**2
    Fv1 = 8 * p["mu_v"] / (r_h_v**2 * A_v * p["rho_v"] * p["lam"])
    Fv3 = (0.019 * p["mu_v"] / (A_v * r_h_v**2 * p["rho_v"] * p["lam"])) * Re_v**0.75
    F_v = np.select([lam_ & lowM, lam_ & ~lowM, ~lam_ & lowM, ~lam_ & ~lowM],
                    [Fv1, Fv1 * comp**-0.5, Fv3, Fv3 * comp**-0.75])
    o.update(Re_v=Re_v, M_v=M_v, F_v=F_v, laminar=lam_, low_mach=lowM)
    o["dPv_dx"] = -(F_v * Q)

    # ---- Kernel 7: capillary limit, Eqs. (2-58a), (2-62)-(2-65) -------
    L_eff = 0.5 * hp.Lc + hp.La + 0.5 * hp.Le
    o["L_eff"] = L_eff
    o["P_cm_e"] = o["P_cm"] - p["rho_l"] * hp.g * dv * np.cos(hp.psi)
    axial_g = p["rho_l"] * hp.g * hp.Lt * np.sin(hp.psi)     # gravity-assisted (+)
    o["Qc"] = (o["P_cm_e"] + axial_g) / (o["F_l"] + F_v) / L_eff

    # ---- Kernel 8: effective wick conductivity, Eq. (2-48) ------------
    kl = p["k_l"]
    o["ke"] = kl * ((kl + k_wick) - (1 - eps) * (kl - k_wick)) / \
                   ((kl + k_wick) + (1 - eps) * (kl - k_wick))

    # ---- Kernel 9: resistance network, Eqs. (2-81), (2-82) ------------
    r_o, r_i = hp.do / 2.0, hp.di / 2.0
    t_p = (hp.do - hp.di) / 2.0
    t_w = hp.T_w * n_layers
    A_hp = np.pi * r_o**2
    o.update(r_o=r_o, r_i=r_i, A_hp=A_hp)
    R = {
        "R_p,e": r_o * t_p / (2 * hp.Le * k_wall),
        "R_w,e": r_o**2 * t_w / (2 * hp.Le * r_i * o["ke"]),
        "R_v"  : np.pi * r_o**2 * F_v * (hp.Le/6 + hp.La + hp.Lc/6) * T
                 / (p["rho_v"] * p["lam"]),
        "R_w,c": r_o**2 * t_w / (2 * hp.Lc * r_i * o["ke"]),
        "R_p,c": r_o * t_p / (2 * hp.Lc * k_wall),
    }
    R["total"] = sum(R.values())                 # area-based [m^2*K/W]
    o["R"]  = R
    o["U"]  = 1.0 / R["total"]                   # [W/(m^2*K)]
    o["dT"] = Q * R["total"] / A_hp              # [K]

    # ---- Kernel 10: sonic limit, Eq. (3-10) ---------------------------
    o["Qs"] = A_v * p["rho_v"] * p["lam"] * np.sqrt(
        p["gamma"] * p["R_v"] * T / (2.0 * (p["gamma"] + 1.0)))

    # ---- Kernel 11: entrainment limit, Eq. (3-17) ---------------------
    o["Qe"] = A_v * p["lam"] * np.sqrt(p["sigma"] * p["rho_v"] / (2.0 * (W_s / 2.0)))

    # ---- Kernel 12: boiling limit, Eq. (3-24) -------------------------
    r_v = dv / 2.0
    o["Qb"] = (2.0 * np.pi * hp.Le * o["ke"] * T) / \
              (p["lam"] * p["rho_v"] * np.log(r_i / r_v)) * \
              (2.0 * p["sigma"] / r_n_agr - o["P_cm_e"])

    # ---- Kernel 13: viscous limit (Busse) -----------------------------
    o["Qv"] = A_v * r_v**2 * p["lam"] * p["rho_v"] * p["P_v"] / (16.0 * p["mu_v"] * L_eff)

    # ---- Kernel 14: governing limit -----------------------------------
    lims = np.vstack([o["Qc"], o["Qs"], o["Qe"], o["Qb"], o["Qv"]])
    o["limits"] = lims
    o["Qgov"] = np.min(lims, axis=0)
    o["gov"]  = np.array([lim_names_agr[int(np.argmin(c))] for c in lims.T])

    # ---- Kernel 15: axial conduction through the wall -----------------
    A_wall = np.pi * (r_o**2 - r_i**2)
    o["Qax"] = k_wall * A_wall * o["dT"] / hp.Lt
    return o

# ======================================================================
# RUN THE MODEL (design wick-layer count + kernel-16 sweep)
# ======================================================================
res_agr   = heat_pipe_model_agr(hp.n_wick_layers)
sweep_agr = {n: heat_pipe_model_agr(n) for n in n_values_agr}

# ======================================================================
# PRINTED TABLES
# ======================================================================
rows_agr = [j for j in range(len(T_agr))
           if np.isclose((T_agr[j] - T_agr[0]) % T_step_agr, 0) or j == len(T_agr) - 1]

def table_agr(title, cols):
    """cols: list of (header, array, format_spec)."""
    print("\n" + "=" * 78)
    print(title)
    print("=" * 78)
    head = f"{'T [K]':>8}" + "".join(f"{h:>15}" for h, *_ in cols)
    print(head)
    print("-" * len(head))
    for j in rows_agr:
        line = f"{T_agr[j]:8.0f}"
        for _, a, spec in cols:
            v = a[j]
            line += f"{v:>15}" if isinstance(v, str) else f"{format(v, spec):>15}"
        print(line)

print("W-Re / SILVER HEAT PIPE (kernels 4-16) completed.")
print("-" * 78)
print(f"Wall                     : W-Re")
print(f"Wick                     : {'W-Re' if WICK_IS_WRE_agr else 'pure W'}")
print(f"Working fluid            : silver")
print(f"Temperature range        : {T_agr[0]:.0f} K to {T_agr[-1]:.0f} K, step {T_INC_agr:.0f} K")
print(f"Operating heat load Q    : {hp.Q} W")
print(f"Capillary radius r_c     : {res_agr['r_c']:.4e} m")
print(f"Porosity / permeability  : {res_agr['eps']:.4f} / {res_agr['K']:.4e} m^2")
print(f"Vapor-core diameter dv   : {res_agr['dv']:.4e} m")
print(f"Effective length L_eff   : {res_agr['L_eff']:.4f} m")
print(f"Chi reference area pi*r_o^2 : {res_agr['A_hp']:.4e} m^2")
print(f"W-Re k                   : EXTRAPOLATED at {np.sum(extrapolated_agr)} of {len(T_agr)} points "
      f"(scaled pure-W curve, ratio {k_ratio_agr:.4f} at 1353 K)")
print("NOTE: silver mu_l, k_l and mu_v are PROVISIONAL; rho_l, sigma and mu_l")
print("      are extrapolated above ~1570-1770 K (see silver_properties.py).")

table_agr("K4  Maximum capillary pressure",
         [("P_cm [Pa]", res_agr["P_cm"], ".4e")])
table_agr("K5  Liquid flow",
         [("dPl/dx [Pa/m]", res_agr["dPl_dx"], ".4e"),
          ("V_l [m/s]",     res_agr["V_l"],    ".4e"),
          ("F_l",           res_agr["F_l"],    ".4e")])
table_agr("K6  Vapor flow",
         [("Re_v",          res_agr["Re_v"],   ".2f"),
          ("M_v",           res_agr["M_v"],    ".4f"),
          ("F_v",           res_agr["F_v"],    ".4e"),
          ("dPv/dx [Pa/m]", res_agr["dPv_dx"], ".4e")])
table_agr("K7  Capillary limit",
         [("P_cm,e [Pa]", res_agr["P_cm_e"], ".4e"),
          ("Qc [W]",      res_agr["Qc"],     ".2f")])
table_agr("K8  Conductivities [W/(m*K)]",
         [("k_WRe",    k_WRe_agr,       ".3f"),
          ("k_l (Ag)", Ag_agr["k_l"],   ".3f"),
          ("k_e",      res_agr["ke"],   ".3f")])
table_agr("K9  Resistances (Chi area-based) [m^2*K/W]",
         [(key, res_agr["R"][key], ".3e") for key in ["R_p,e", "R_w,e", "R_v", "R_w,c", "R_p,c"]])
table_agr("K9  Total resistance, U and temperature drop",
         [("R_tot", res_agr["R"]["total"], ".3e"),
          ("U [W/m2K]", res_agr["U"],      ".3e"),
          ("dT [K]",    res_agr["dT"],     ".4f")])
table_agr("K10-K13  Operating limits [W]",
         [("Qs", res_agr["Qs"], ".3e"), ("Qe", res_agr["Qe"], ".3e"),
          ("Qb", res_agr["Qb"], ".3e"), ("Qv", res_agr["Qv"], ".3e")])
table_agr("K14 Governing limit",
         [("Qgov [W]", res_agr["Qgov"], ".3e"),
          ("Governing", res_agr["gov"], ""),
          ("Margin [W]", res_agr["Qgov"] - hp.Q, ".3e")])
table_agr("K15 Axial wall conduction",
         [("Qax [W]", res_agr["Qax"], ".4e"),
          ("% of Q",  res_agr["Qax"] / hp.Q * 100, ".4f")])

# ---- K14 governing ranges ----------------------------------------------
print("\nK14 Governing-limit ranges:")
print(f"  {'Limit':<12} {'T start':>8} {'T end':>8} {'Q start [W]':>13} {'Q end [W]':>13}")
start_agr = 0
for j in range(1, len(T_agr) + 1):
    if j == len(T_agr) or res_agr["gov"][j] != res_agr["gov"][j - 1]:
        print(f"  {res_agr['gov'][start_agr]:<12} {T_agr[start_agr]:8.0f} {T_agr[j-1]:8.0f} "
              f"{res_agr['Qgov'][start_agr]:13.1f} {res_agr['Qgov'][j-1]:13.1f}")
        start_agr = j

feasible_agr = hp.Q <= res_agr["Qgov"]
if np.all(feasible_agr):
    print(f"\nOperating Q = {hp.Q} W is FEASIBLE over {T_agr[0]:.0f}-{T_agr[-1]:.0f} K.")
else:
    print(f"\nOperating Q = {hp.Q} W EXCEEDS the governing limit at "
          f"{np.sum(~feasible_agr)} of {len(T_agr)} points "
          f"({T_agr[~feasible_agr].min():.0f}-{T_agr[~feasible_agr].max():.0f} K).")

# ---- K16 sweep table ----------------------------------------------------
print("\n" + "=" * 78)
print("K16 Wick-layer sweep: governing limit")
print("=" * 78)
print(f"{'n':>3} {'T [K]':>7} {'Qgov [W]':>12} {'Governing':>12}")
for n in n_values_agr:
    for j in rows_agr:
        print(f"{n:>3} {T_agr[j]:7.0f} {sweep_agr[n]['Qgov'][j]:12.3e} {sweep_agr[n]['gov'][j]:>12}")

# ======================================================================
# PLOTS — one independent figure per result
# ======================================================================
sub_agr = (f"W-Re wall, {'W-Re' if WICK_IS_WRE_agr else 'W'} wick, silver fluid; "
           f"W-Re k extrapolated > 1353 K")

def plot_agr(y, ylabel, title, color, log=False, absval=False, Qline=False):
    y = np.abs(y) if absval else y
    plt.figure(figsize=(9, 6))
    plt.plot(T_agr, y, '-', color=color, lw=2)
    if Qline:
        plt.axhline(hp.Q, color='k', ls='--', lw=2, label=f'Operating Q = {hp.Q} W')
        plt.legend()
    plt.xlabel("Temperature, T [K]")
    plt.ylabel(ylabel)
    if log:
        plt.yscale('log')
    plt.title(f"{title}\n({sub_agr})")
    plt.grid(True, which='both', alpha=0.4)
    plt.tight_layout()
    plt.show()

# K4
plot_agr(res_agr["P_cm"], "Maximum Capillary Pressure, P_cm [Pa]",
        "Maximum Capillary Pressure", 'tab:blue')
# K5
plot_agr(res_agr["dPl_dx"], "Liquid Pressure Gradient, dP_l/dx [Pa/m]",
        "Liquid Pressure Gradient", 'tab:cyan')
plot_agr(res_agr["V_l"], "Liquid Velocity, V_l [m/s]",
        "Liquid Velocity in Wick", 'tab:cyan')
# K6
plot_agr(res_agr["dPv_dx"], "|Vapor Pressure Gradient|, |dP_v/dx| [Pa/m]",
        "Vapor Pressure Gradient", 'tab:gray', log=True, absval=True)
plot_agr(res_agr["Re_v"], "Vapor Reynolds Number, Re_v [-]",
        "Vapor Reynolds Number", 'tab:gray', log=True)
plot_agr(res_agr["M_v"], "Vapor Mach Number, M_v [-]",
        "Vapor Mach Number", 'tab:gray', log=True)
# K7
plot_agr(res_agr["Qc"], "Heat Transport Rate, Q [W]",
        "Capillary Limit Qc", 'tab:red', Qline=True)

# K8
plt.figure(figsize=(9, 6))
plt.plot(T_agr, k_WRe_agr,     'k-',  lw=2, label='$k$ W-Re (extrapolated > 1353 K)')
plt.plot(T_agr, Ag_agr["k_l"], ':',   color='tab:cyan',   lw=2, label='$k_l$ liquid silver (provisional)')
plt.plot(T_agr, res_agr["ke"], '--',  color='tab:purple', lw=2, label='$k_e$ wick (metal + Ag)')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Thermal Conductivity [W/(m·K)]")
plt.title(f"Effective Wick Thermal Conductivity vs. Its Constituents\n({sub_agr})")
plt.grid(True, alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()

# K9: resistance components (K/W)
R_cols_agr = {"R_p,e": 'tab:blue', "R_w,e": 'tab:green', "R_v": 'tab:gray',
             "R_w,c": 'tab:olive', "R_p,c": 'tab:brown'}
R_tex_agr  = {"R_p,e": '$R_{p,e}$ (evap. wall)', "R_w,e": '$R_{w,e}$ (evap. wick)',
             "R_v": '$R_v$ (vapor)', "R_w,c": '$R_{w,c}$ (cond. wick)',
             "R_p,c": '$R_{p,c}$ (cond. wall)'}
plt.figure(figsize=(9, 6))
for key_agr, c_agr in R_cols_agr.items():
    ls_agr = ':' if key_agr in ("R_w,c", "R_p,c") else '-'
    plt.plot(T_agr, res_agr["R"][key_agr] / res_agr["A_hp"], ls_agr, color=c_agr,
             lw=3 if ls_agr == ':' else 2, label=R_tex_agr[key_agr])
plt.plot(T_agr, res_agr["R"]["total"] / res_agr["A_hp"], 'k-', lw=2.5, label='$R_{total}$')
plt.xlabel("Temperature, T [K]")
plt.ylabel("Thermal Resistance [K/W]")
plt.yscale('log')
plt.title(f"Heat Pipe Resistance Network Components vs. Temperature\n({sub_agr})")
plt.grid(True, which='both', alpha=0.4)
plt.legend(fontsize=9)
plt.tight_layout()
plt.show()

plot_agr(res_agr["dT"], "Total Temperature Drop, ΔT [K]",
        f"Total Temperature Drop at Q = {hp.Q} W", 'tab:red', log=True)
# K10-K13
plot_agr(res_agr["Qs"], "Heat Transport Rate, Q [W]", "Sonic Limit Qs",
        'tab:blue', log=True, Qline=True)
plot_agr(res_agr["Qe"], "Heat Transport Rate, Q [W]", "Entrainment Limit Qe",
        'tab:green', Qline=True)
plot_agr(res_agr["Qb"], "Heat Transport Rate, Q [W]", "Boiling Limit Qb",
        'tab:gray', log=True, Qline=True)
plot_agr(res_agr["Qv"], "Heat Transport Rate, Q [W]", "Viscous Limit Qv",
        'tab:orange', log=True, Qline=True)

# K14: all limits + governing envelope
lim_cols_agr = ['tab:red', 'tab:blue', 'tab:green', 'tab:gray', 'tab:orange']
plt.figure(figsize=(10, 7))
for nm_agr, c_agr, q_agr in zip(lim_names_agr, lim_cols_agr, res_agr["limits"]):
    plt.plot(T_agr, q_agr, '--', color=c_agr, lw=1.5, alpha=0.7, label=nm_agr)
plt.plot(T_agr, res_agr["Qgov"], 'k-', lw=3, alpha=0.6, label='Governing limit')
plt.axhline(hp.Q, color='k', ls=':', lw=2, label=f'Operating Q = {hp.Q} W')
plt.xlabel("Temperature, T [K]", fontsize=12)
plt.ylabel("Heat Transport Limit, Q [W]", fontsize=12)
plt.yscale('log')
plt.title(f"Governing Heat Pipe Limit vs. Temperature (5 limits)\n({sub_agr})", fontsize=13)
plt.grid(True, which='both', alpha=0.4)
plt.legend(fontsize=9)
plt.tight_layout()
plt.show()

# K15
plot_agr(res_agr["Qax"] / hp.Q * 100, f"Axial Wall Conduction [% of Q = {hp.Q} W]",
        "Axial Heat Conduction Through the Pipe Wall", 'tab:brown', log=True)

# K16: wick-layer sweep, one figure per limit + governing
n_cols_agr = plt.cm.viridis(np.linspace(0, 1, len(n_values_agr)))
for key_agr, ttl_agr in [("Qc", "Capillary Limit"), ("Qs", "Sonic Limit"),
                       ("Qe", "Entrainment Limit"), ("Qb", "Boiling Limit"),
                       ("Qv", "Viscous Limit"), ("Qgov", "Governing Limit")]:
    plt.figure(figsize=(9, 6))
    for n, c_agr in zip(n_values_agr, n_cols_agr):
        plt.plot(T_agr, sweep_agr[n][key_agr], '-', color=c_agr, lw=2, label=f'n = {n} layers')
    plt.axhline(hp.Q, color='k', ls='--', lw=1.5, label=f'Operating Q = {hp.Q} W')
    plt.xlabel("Temperature, T [K]")
    plt.ylabel("Heat Transport Rate, Q [W]")
    plt.yscale('log')
    plt.title(f"{ttl_agr} vs. Wick Layer Count\n({sub_agr})")
    plt.grid(True, which='both', alpha=0.4)
    plt.legend(fontsize=9)
    plt.tight_layout()
    plt.show()
```

```python
# ============================================================================
# KERNEL 20 — SIMPLE COMPARISON PLOTS: Li vs Ag, W vs W-Re
# ============================================================================
# Overlays the four cases already computed in kernels 17-19:
#   Lithium + W     (kernel 17, W_cmp)      Lithium + W-Re  (kernel 17, WRe_cmp)
#   Silver  + W     (kernel 18, res_ag)     Silver  + W-Re  (kernel 19, res_agr)
# Run kernels 17, 18 and 19 first. Nothing is recomputed or overwritten.
#
# Colors: blue = lithium, orange = silver.  Line style: dashed = W, solid = W-Re.
# Lithium uses hp.temperature_range(); silver uses its own hotter range, so
# the two fluids sit side by side on the temperature axis.
# ============================================================================
for _name_cmp20 in ("W_cmp", "WRe_cmp", "T_cmp", "res_ag", "T_ag", "res_agr", "T_agr"):
    if _name_cmp20 not in globals():
        raise NameError(f"{_name_cmp20} not found - run kernels 17, 18 and 19 first.")

cases_cmp20 = [
    ("Li + W",    T_cmp, W_cmp,   'tab:blue',   '--'),
    ("Li + W-Re", T_cmp, WRe_cmp, 'tab:blue',   '-'),
    ("Ag + W",    T_ag,  res_ag,  'tab:orange', '--'),
    ("Ag + W-Re", T_agr, res_agr, 'tab:orange', '-'),
]

def compare_plot_cmp20(key, ylabel, title, log=True, Qline=True, absval=False):
    plt.figure(figsize=(10, 6))
    for label, T, res, color, ls in cases_cmp20:
        y = np.abs(res[key]) if absval else res[key]
        plt.plot(T, y, ls, color=color, lw=3 if ls == '--' else 1.8,
                 alpha=0.85, label=label)
    if Qline:
        plt.axhline(hp.Q, color='k', ls=':', lw=2, label=f'Operating Q = {hp.Q} W')
    plt.xlabel("Temperature, T [K]")
    plt.ylabel(ylabel)
    if log:
        plt.yscale('log')
    plt.title(title)
    plt.grid(True, which='both', alpha=0.4)
    plt.legend()
    plt.tight_layout()
    plt.show()

# 1. Governing limit
compare_plot_cmp20("Qgov", "Heat Transport Limit, Q [W]",
                   "Governing Heat Pipe Limit: Lithium vs Silver, W vs W-Re")

# 2-6. Individual limits
compare_plot_cmp20("Qc", "Heat Transport Rate, Q [W]", "Capillary Limit Qc")
compare_plot_cmp20("Qs", "Heat Transport Rate, Q [W]", "Sonic Limit Qs")
compare_plot_cmp20("Qe", "Heat Transport Rate, Q [W]", "Entrainment Limit Qe")
compare_plot_cmp20("Qb", "Heat Transport Rate, Q [W]", "Boiling Limit Qb")
compare_plot_cmp20("Qv", "Heat Transport Rate, Q [W]", "Viscous Limit Qv")

# 7. Total temperature drop
compare_plot_cmp20("dT", "Total Temperature Drop, ΔT [K]",
                   f"Total Temperature Drop at Q = {hp.Q} W", Qline=False)

# 8. Maximum capillary pressure
compare_plot_cmp20("P_cm", "Maximum Capillary Pressure, P_cm [Pa]",
                   "Maximum Capillary Pressure", log=False, Qline=False)

# 9. Effective wick conductivity
compare_plot_cmp20("ke", "Effective Wick Conductivity, k_e [W/(m·K)]",
                   "Effective Wick Thermal Conductivity", log=False, Qline=False)
```
